###새롭게 알게된 것
1. 너의 부모를 가져와라!
부모 기준으로 조인
left join 부모 테이블 on 자식 테이블
자식의 부모아디이가 부모아이디와 같다! 

2. left 조인으로 저건쓰지 말고
그냥 join으로 아이디 같은거만 가져와도 괜찮다
부모 없는 데이터는 애초에 만족할 ㅅ ㅜ없으니까


```python
SELECT child.ID, child.GENOTYPE, parent.GENOTYPE as PARENT_GENOTYPE
-- 부모 대장균 형질 
FROM ECOLI_DATA parent
join ECOLI_DATA child 
on parent.id = child.parent_id
where child.GENOTYPE & parent.GENOTYPE = parent.GENOTYPE
ORDER BY child.ID
```
