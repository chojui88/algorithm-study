### 질문
1. select 문에, 여러가지 칼럼 정보를 한 칼럼으로 산출하는건 어떻게해?
-> concat(a,b)

2. where 절에서, 값이 가장 큰 것 하나만 산출하려면 어떻게 해?
-> max() 대신 where 절에서 쓸 수 없으므로, select 내부 절 만들어서 사용해야함

3. where절 안의 서브퀄리에서도 as 표기 해야하나? b.view 이렇게?
-> 아니!

```python
SELECT concat('/home/grep/src/',f.BOARD_ID,f.FILE_NAME,f.FILE_EXT) as FILE_PATH
from USED_GOODS_BOARD b 
join USED_GOODS_FILE f
on b.BOARD_ID = f.BOARD_ID
where max(b.VIEWS)

-- max는 집계함수라서 where 절에서 사용할수 없다
```

```python
SELECT concat('/home/grep/src/',f.BOARD_ID,f.FILE_NAME,f.FILE_EXT) as FILE_PATH
from USED_GOODS_BOARD b 
join USED_GOODS_FILE f
on b.BOARD_ID = f.BOARD_ID
where max(b.VIEWS)
```

```python
-- 코드를 입력하세요
SELECT concat('/home/grep/src/',f.BOARD_ID,f.FILE_NAME,f.FILE_EXT) as FILE_PATH
from USED_GOODS_BOARD b 
join USED_GOODS_FILE f
on b.BOARD_ID = f.BOARD_ID
where ( select max(VIEWS) -- 여기서 어떤 값이랑 비교를 하는지가 없음
    from USED_GOODS_BOARD)
group by BOARD_ID
order by FILE_PATH desc
```

```python
SELECT concat('/home/grep/src/',f.BOARD_ID,'/',f.FILE_ID,f.FILE_NAME,f.FILE_EXT) as FILE_PATH
from USED_GOODS_BOARD b 
join USED_GOODS_FILE f
on b.BOARD_ID = f.BOARD_ID
where b.views = ( select max(VIEWS)
    from USED_GOODS_BOARD)
order by FILE_PATH desc -- 이 정렬은 넣어줘야 함!
```
