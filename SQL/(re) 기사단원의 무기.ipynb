```python
#1차시도

def solution(number, limit, power):
    #number : 기사단원의 수 
    # limit: 정해진 공격력
    # power: 제한수치 초과한 기사가 사용할 무기 공격력
    
    #1~numver 까지 번호
    # 기사 번호의 약수 개수에 해당 하는 공격력 가진 무기
    #이웃나라와 협약
    # 공격력 제한수치 < 공격력
    # 협약기관에서 정한 공격력 가지는 무기
    
    answer = []
    for i in range(1,number+1): #만약 4라면
        sum = 0
        for j in range(1,i**0.5):
            if i%j == 0: #4 나누기 1, 2, 4
                sum+=1

            sum *= 2

            if (i**0.5)**2 == i:
                sum -=1
            
    #1. 약수의 개수 세고
    #2. 약수의 개수만큼 공격력 가진 무기 구입, 제한보다 작은 kg으로 
        
        if limit >= sum:
            answer.append(sum)
        else:
            answer.append(limit-1)
    
    #15번 기상단원은 15의 약수가 4개이므로 4개
    # 무기점 주인이 모두 무기 만들기 위해 필요한 철의 무게
    print(answer)
    res = 0
    for i in answer:
        res+=i
    return res
```

```python
#2차 시도

def solution(number, limit, power):
    #number : 기사단원의 수 
    # limit: 정해진 공격력
    # power: 제한수치 초과한 기사가 사용할 무기 공격력
    
    #1~numver 까지 번호
    # 기사 번호의 약수 개수에 해당 하는 공격력 가진 무기
    #이웃나라와 협약
    # 공격력 제한수치 < 공격력
    # 협약기관에서 정한 공격력 가지는 무기
    
    answer = []
    for i in range(1,number+1): #만약 4라면
        sum = 0
        for j in range(1,int(i**0.5)+1): #1부터 3까지!
            if i%j == 0: #4 나누기 1, 2, 4
                sum+=1

        sum *= 2

        if (i**0.5)**2 == i:
                sum -=1
            
    #1. 약수의 개수 세고
    #2. 약수의 개수만큼 공격력 가진 무기 구입, 제한보다 작은 kg으로 
        
        if limit >= sum:
            answer.append(sum)
        else:
            answer.append(power)
    
    #15번 기상단원은 15의 약수가 4개이므로 4개
    # 무기점 주인이 모두 무기 만들기 위해 필요한 철의 무게
  
    res = 0
    for i in answer:
        res+=i
    return res
```
