```python
# 모든 찾기
# 1초
# Sho 어레이 a 를 가지고 있고
# n의 정수를 가지고 있는대

# 2개의 다른 i랑 j그리고 ak, aj를 지웠대!

# 대신 여기서 말ㄹ하는 숫자는 인덱스 + 1 

#동작을 한뒤에, 구별적인 요수들이 있는데
# 가장 가능한 가장 긴! 것을 원해

# 입력값 , 50 이하 
t = int(input())

for _ in range(t):
    n = int(input()) #배열의 길이
    a_i = list(map(int, input().split())) #리스트의 값


    #1. 일단 정렬먼저
    a_i = sorted(a_i)

    count = 0
    #2. for 문 돌려서 각각의 값이 가지고 있는 1번째 값이 중복되고 있는 개수가 있는지 확인해서
    for i in range(n):
        for j in range(i+1,n):
            if a_i[i] == a_i[j]:
                count +=1
            else:
                break
    #3. 이중 for문으로 가지고 있는 개수가 몇개인지 확인해서 카운트에 담기
    #4. 다음 for문에서 또 중복값 찾기

    #5. 외종적으로 count의 개수가 %2 == 0 이면 len(list) - count 하면 되고
    if count%2 == 0:
        print(len(a_i) - count)
    else:
        print(len(a_i) - count + 1)
    #5-1 만약 아니라면, len(list) - count + 1 하면 된다


#결과값
# 하나의 정수를 도출해야한느데, 최종 array의 길이야
# 

```
