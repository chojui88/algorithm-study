```python
#문제를 풀면 풍선 1개를 받고
#팀이 문제를 해결할 때마다, 풍선 1개 더 받는다
#첫번째 문제 푼 사람 1개 더 받는다

#콘테스트는 26개의 문제가 있다
#a~z 까지

#순서대로 문제를 해결해야하는데
#s string 이 주어지고 이것은 문제들이 해결된  순서대로 
# 아무 팀도 문제를 2번 해결할 수 없다

#팀들이 받은 풍선의 총 개수는? 
#그리고 기록하시오, 같은 문제가 아마 해결돤, 아무 팀에도 아닌?

#처음으로 해결한팀은 풍선 2개

#이미 다른 팀이 해결한 문제 푼 팀은 1개 
#s 해결된 순서

t = int(input())

#1. 각각의 테스트 케이스는 n숫자를 포함한다 ( string의 길이)
#2. 3번째 줄 부터는 n의 길이로 구성된 대문자 영어 문자, denoting 순서대로 문제를 해결하라 

# 하나의 숫자 결과, 최종 수의 풍선은 팀이 받는것이다 
for _ in range(t):
    n = int(input())
    s = input()

    seen = set()
    answer = 0

    for char in s:
        if char not in seen:
            answer +=2 
            seen.add(char)
        else: 
            answer +=1 

        #각가의 테스트케이스마다! 하나의 정수로, 그 팀들이 받은 풍선의 개수를 출력하시오
    print(answer)
```

```python
t = int(input())

for _ in range(t):
    n = int(input())
    s = [input() for _ in range(n)]

    set_s = set(s)
    answer = ""

    for word in s:
        check = 0

        for i in range(1, len(word)):
            a = word[:i]
            b = word[i:]

            if a in set_s and b in set_s:
                check = 1
                break

        answer += str(check)

    print(answer)
```
