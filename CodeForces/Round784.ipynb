```python
# array a -> n 
# value -> three times / -1 배열에서 3번 이상 등장하면, 그 숫자 하나 출력

from collections import Counter

t = int(input())

# 배열의 길이 n

for _ in range(t):
    n = int(input())
    a = list(map(int, input().split()))

    count = Counter(a) #배열 

    answer = -1

    for num, cnt in count.items():
        if cnt >= 3:
            answer = num
            break

    print(answer)



        
    
    

```
