```python
def compare(arr1,arr2):
    if sum(arr1)>sum(arr2):
        return 1
    elif sum(arr1)<sum(arr2):
        return -1
    else:
        if len(arr1)>len(arr2):
            return 1
        elif len(arr1)<len(arr2):
            return -1
        else: return 0

print(compare([10, 20, 30], [50, 10]))
print(compare([1, 2, 3], [10, 20]))
print(compare([10, 10], [5, 5, 5, 5]))
print(compare([1, 2, 3], [1, 2, 3]))
```
