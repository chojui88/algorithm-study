```python
def f_a():
    num = 20
    print("f_a()의 num값 %d"% num)
def f_b():
    print("f_b()의 num값 %d"%num)

num = 10 #w
f_a()
f_b()
```

최상단 레벨에서 n=10 이라고 선언되었다
-> 전역변수, 어디든 접근 가능

f_a() 내부 지역 변수 선언
    이 지역변수는 외부 전역변수와 별계
    그래서 내부에서 추력하면 지역변수 값 출력
    호출후 사라진다

f_b() 내부에 할당 문장이 없다
    첫번째, 함수 내부의 지역 변수 찾는다
    없으므로
    두 번쨰, 전역 스코프에서 num 변수를찾는다
    있으므로 사용한다

```python
def f_a():
    global num
    num = 20
    print("f_a()의 num값 %d" %num)

num = 10
f_a()
print("global num %d" % num)
```

전역변수 num 생성
global 키워드, 함수 외부에 전역변수 참조하겠다 선언
num = 10 전역변수의 값을 변경한단ㄴ 뜻!!!!!
결국 값이 변경됨!!!!!!

```python
def calc(v1,v2,op):
    result = 0
    if op =='+':
        result = v1+v2
    elif op == '-':
        result = v1-v2
    elif op == '/':
        if v2 == 0:
            return 'error'
        else:
            return v1 / v2
    return result



while True:
    oper = input("연산자를 입력하세요:")
    if oper == 'q':
        print("프로그램을 종료합니다.")
        break
    var1 = int(input("첫 번째 수를 입력:"))
    var2 = int(input("두 번째 수를 입력:"))
    res = calc(var1,var2,oper)
    if res == 'error':
        print("0으로 나누면 안됩니다!")
    else:
        print("## 계산기 %d %s %d = %d" % (var1,oper,var2,res))
```

```python

from functools import reduce
def sum_all(a,b):
    all = range(a,b+1)
    return reduce(lambda a,b:a+b, all)
print("0 to 100:",sum_all(0,100))
print("0 to 1000:", sum_all(0,1000))
print("50 to 1000:", sum_all(50,1000))
```

```python
def solution(arr1, arr2):
    len1 = len(arr1)
    len2 = len(arr2)

    if len1 != len2:
        return -1 if len1>len2 else 1
    else:
        sum1 = sum(arr1)
        sum2 = sum(arr2)
        if sum1<sum2:
            return -1 if sum1>sum2 else 1
        else:
            return 0

print(solution([49, 13], [70, 11, 2]))
print(solution([100, 17, 84, 1], [55, 12, 65, 36]))
print(solution([1, 2, 3, 4, 5], [3, 3, 3, 3, 3]))
```
