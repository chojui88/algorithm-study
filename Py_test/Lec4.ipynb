```python
# 연습문제 2 풀이
total_sum = 0      # 3의 배수들의 합을 저장할 변수 초기화
current_number = 1 # 1부터 1000까지 증가할 숫자를 저장할 변수 초기화

while current_number <= 1000: # current_number가 1000 이하인 동안 반복
    if current_number % 3 == 0:  # current_number가 3의 배수인지 확인
        total_sum += current_number  # 3의 배수이면 total_sum에 더함
    current_number += 1          # 다음 숫자로 1 증가

print(f'1부터 1000까지의 자연수 중 3의 배수의 합은 {total_sum}입니다.')
```

```python
from functools import reduce
def sum_all(a,b):
    all = range(a,b+1)
    return reduce(lambda x,y : x+y,all)

print(sum_all(0,100))
```
