```python
import random

visitors = []

for i in range(100):
    visitors.append(random.randint(0,100))
   
ages = {
    "baby":0,
    "child":0,
    "teen":0,
    "adult":0,
    "senior":0
}
price = {
    "baby":0,
    "child":200,
    "teen":300,
    "adult":500,
    "senior":0
}


for i in range(100):
    v = visitors.pop()
    if v <=7:
        ages['baby']+=1
    elif v<=13:
        ages['child']+=1
    elif v<=19:
        ages['teen']+=1
    elif v<=64:
        ages['adult']+=1
    else:
        ages['senior']+=1

for i in ages:
    sum += ages[i] * price[i]
print(f"영유아: {ages['baby']}명 {ages['baby']*price['baby']}원")
print(f"어린이: {ages['child']}명 {ages['child']*price['child']}원")
print(f"청소년: {ages['teen']}명 {ages['teen']*price['teen']}원")
print(f"성인: {ages['adult']}명 {ages['adult']*price['adult']}원")
print(f"어르신: {ages['senior']}명 {ages['senior']*price['senior']}원")
print(f"총 명수 : {sum}")
```
