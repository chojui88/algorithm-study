#### 왜 이 알고리즘?
트리 구조: 판매원 한명의 부모 노드 있음
bottom-up 

#### 시간 복잡도
O(N * M)
n = 전체 판매원수 + 총 판매 건수 x 트리의 높이
판매건마다 트리 타고 위로 올라가므로, 트리 높이를 곱함

#### 공간 복잡도
O(N)
tree 사전과 price 사전에 각각 N개의 정보를 저장

#### 엣지 케이스
1. 상납금이 0원이 되는 경우
//10으로 인해서, 금액이 10원 미만이면, 상납금은 0원이 됨. 소수점만 존재할수도
이때 while을 종료하는게 더이상 불필요한 탐색 막을 수 있음

2. 동일한 판매자가 여러번 물건을 팔 수 있다
price 기존 금액에 누적해서 써야한다

3. 최상위 노드 처리
문제는 enroll에 있는 사람들의 이익만 반환, 
-는 price에 없으므로, 적절히 제외해야 한다

#### 새로 배운 것
1. 상향식 트리 조회
부모로 올라가는 단방향 = parent[chile] = parent 형태의 map으로 구현 가능

2. 정수 몫 연산과 절사 법칙: float를 소수점 연산에 쓰면 오차!!
정수 나눗셈//10을 쓰는것이 가장 좋음

3. 가지치기ㅣ - 0보다 클때만 상승한다는 조건을 추가하여 불필요한 연산 막을 수 있도록 최적화


```python
def solution(enroll, referral, seller, amount):
    #판매원 이름, 그 판매원을 조직에 참여시킨 이름, 
    # 집계 데이터의 판매원 이름 나열
    # 판매 수량

    # 칫술 판매하면, 이익이 피라미드 조직 타고 조금씩 분배
    # 모든 판매원은 칫솔 하나 판매 생기는 이익에서 10%를 자신의 추천인에게 배분, 나머지 90%는 자신이
    
    # seller에 있는, 판매원 이름에서 
    tree = {}
    for en, re in zip(enroll, referral):
        tree[en] = re
   
    price = {}
    for en in enroll:
        price[en] = 0

    money = 0
    for se, am in zip(seller, amount):
            money = (100 * am)
            price[se] += money
            while se != '-' and money>0:
                price[tree[se]] += int(money * 0.1)
                price[se] -= int(money * 0.1)
                se = tree[se]

    # 판매원은 자신의 발생 이익 90% + 내 밑의 이익금 10% 가져간다 여기의 10% 추천인에게 분배
    # 원 단위에서 자름. 소수점 밑으로는 자신이 가져감


    #각 판매원이 얻은 이익금 나열한 배열을 반환
    # enroll 이름 순으로
    return [price[name] for name in enroll] 
```

```python
def solution(enroll, referral, seller, amount):
    parent = dict(zip(enroll, referral))

    total = {name: 0 for name in enroll}

    for i in range(len(seller)):
        money = amount[i] * 100
        cur_name = seller[i]
        while money > 0 and cur_name != '-':
            total[cur_name] += money - money // 10
            cur_name = parent[cur_name]
            money //=10

    return [total[name] for name in enroll]


```
