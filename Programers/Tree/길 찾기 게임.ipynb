#### 왜 이 알고리즘?

#### 시간 복잡도
O(N log N) ~O(N^2)
노드 정렬: N개 노드 정렬하는건 NLogN
트리 : 균형 이루면 logN만큼 비교 , 편향트리면 N^2
전위/후위 순회: O(N)

#### 공간 복잡도
O(N)
노드 객체, 스택, 결과 배열

#### 엣지 케이스
1. 노드 한개만 주어진다 = [[1],[1]] 형태 결과 반환해야 함
2. 편향 트리 , 재귀는 재귀 깊이 초과가발생할 수 있으나, 스택으로 관리하면 괜찮음

#### 새로 배운 것
1. 정렬할때, key = ,,, lambda  해서

2. 재귀 대신 stack으로 처리
3. 후위 순위는, 부모 노드를 스택에서 2번 만나야 함 그래서 상태 플레그 False추가가 필요
4. 2차원 함수 []*2 하면 얕은 복사임

```python
from collections import deque

#1. Node 클래스 정의
class Node:
    def __init__(self, info, num, left=None, right = None):
        self.info = info
        self.num = num
        self.left= left
        self.right = right

    #2. 왼쪽 자식 노드 있는지 확인하는 함수, 존재하면 True
    def has_left(self):
        return self.left is not None

    #3. 오른쪽 자식 노드 있는지
    def has_right(self):
        return self.right is not None

    #4. 이진트리 생성 함수
    def make_BT(nodeinfo):
        nodes = [i for i in range(1, len(nodeinfo)+1)]
         #1. 정렬, y가 크면 우선순위가 높다
            #2. y가 크다면 x가 작은게 우선순위가 높다
        
        #노드 번호를 정렬(1번부터 시작! i-1번 인덱스에서 가져옴)
        nodes.sort(key = lambda x: (nodeinfo[x-1][1], -nodeinfo[x-1][0]),reverse=True)
        #노드 번호는 1번부터 시작하는데, nodeinfo 
        #5. 노드의 번호 리스트 생성

        #(좌표), 번호
        root = None
        for i in range(len(nodes)):
            if root is None:
                root = Node(nodeinfo[nodes[0]-1], nodes[0])
            else:
                parent = root
                node = Node(nodeinfo[nodes[i]-1], nodes[i])


    #6. 부모 노드의 x좌표가 크면 왼쪽으로
    #7. 부모노드의 x좌표가 작거나 같으면 오른쪽으로
                while True:
                    if parent.info[0] > node.info[0]:
                        if parent.has_left():
                            parent = parent.left
                        else:
                            parent.left = node
                            break

                    elif: parent.info[0] <= node.info[0]:
                        if parent.has_right():
                            parent = parent.right
                        else:
                            parent.right = node
                            break
        return root
  
# 8. 전위 순회 함수 = 부모, 왼자식 오른자식
def pre_order(root,answer):
    #현재 바문할 노드, 방문할 노드를 순서대로 담을 리스트
    stack = [root]
    while stack:
        if root is None:
            return 
        answer.append(root.num)
        # 현재 노드의 번호를 결과 리스트에 추가
        pre_order(root.left, answer)
        #왼쪽 자시 방문해서 다시 순회, 왼쪽 서브트리 완전탐색 먼저@
        pre_order(root.right, answer)

    #ㄱoot가 존재하지 않으면 재귀 호출을 종료하고 이전단계로 돌아감
    #Base code 재귀함수 끊어주는 역할
        

def post_order(root,answer):
    stack = [(root, False)]
    #왼쪽 - 오른쪽 - 나 부모 노드를 스택에서 두번 마두침!
    # 자식 노드를 스택에 집어넣기 위해 부모를 거쳐갈때 
    #자식 다 끝내고 자기 방문처리 할때 

    while stack:
        node, visited = stack.pop()
        if node is None:
            continue  #node가 존재하지 않으면, 다음 while로 넘어가기
        #현재 노드에서 번호 추출해서, 결과 리스트에 추가 
        # 현재 노드의 번호를 결과 리스트에 추가
        if visited: #자식들 다 돌고온 True 값이면
            answer[1].append(node.num) #0이 전위 결과 담는 리스트 
        else: 
            stack.append((node, True)) #이거 해야, answer이 저장되지
            stack.append((node.right,False)) #스택은 나중에 들어간것이 먼저 나온다 
            stack.append((node.left,False))

#9. 후위 순회 함수

#10. 이진트리 생성, 결과 반환 함수
def solution(nodeinfo):
    answer = [[],[]]
    root = make_BT(nodeinfo)

    # 이진트리 배열, 1번 노드부터
    # 루트 노드부터 자식 노드를 재귀로 찾아, 트리를 구축해야한다
    # 방문할 곳의 2차원 좌표 = 이진트리의 노드 
    # 순회 방법을 힌트로! 
    # 좌표로 이진트리를 구축! 
    pre_order(root, answer)
    post_order(root, answer)
    return answer


    
   
   

    #전위, 후위
    
    #x 가로 y 세로

    #전위, 후위 결과를 2차원 배열에 순서대로 담기
    #순회 마친 팀이 승리!
    answer = [[]]
    return answer
```
