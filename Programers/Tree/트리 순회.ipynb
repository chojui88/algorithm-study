#### 왜 이 알고리즘?

#### 시간 복잡도
노드의 개수 N
O(N)
모든 노드를 한번씩만 방문@

#### 공간 복잡도
O(H) 트리의 높이
재귀 호출 스택: 트리 높이만큼, 메모리 스택에 쌓임!
완전 이진 트리 : 높이가 log2N 이므로 Log N 공간 차지@
편향 트리 :O(N) 의 메모리

#### 엣지 케이스
1. 빈 트리 nodes = []
idx = 0 에서 len(nodes) 범위 벗어남
2. 노드 하나만 있는 트리 
왼쪽/오른쪽 재귀 호출에서 base code 에서 걸려 곧바로 탈출
3. 한쪽만 있는 편향트리

#### 새로 배운 것
1. 배열 기반 이진트리 규칙
0번 시작 기준
왼쪽 2 * idx + 1
오른쪽  2 * idx + 2

2. 재귀 함수, 독립 변수
return 으로 새로 만들어진 변수값 인출해서
부모가 이어붙이는 += 방식으로 누적!



```python
def preorder(nodes,idx):
    #이진 트리 리스트
    #idx가 노드 리스크 길이보다 작을때 -> 기저조건 정하기
    if idx < len(nodes):
        #루트 노드를 출력, 왼쪽 오른쪽 재귀호출 -> 출력 순서대로 이어붙임

        ret = str(nodes[idx]) + " "
        #배열에 노드 값을 가져온다 
        ret += preorder(nodes,2*idx)
        ret += preorder(nodes,2*idx + 1)
        return ret

    else:
        return 

        
def inorder(nodes,idx):
    if idx < len(nodes):
        
        ret = inorder(nodes,2*idx)
        #루트를 출력
        ret += str(nodes[idx]) + " "
        ret += inorder(nodes,2*idx + 1)
        return ret

    else:
        return 

def postorder(nodes,idx):
    if idx < len(nodes):
        ret += postorder(nodes,2*idx)
        ret += postorder(nodes,2*idx + 1)
        ret += str(nodes[idx]) + " "
        
    
    else:
            return 


def solution(nodes):

    return[
        preorder(nodes,0)[:-1],
        inorder(nodes,0)[:-1],
        postorder(nodes,0)[:-1]
    ]
        #전위, 중위, 후위 순회 결과 반환
```
