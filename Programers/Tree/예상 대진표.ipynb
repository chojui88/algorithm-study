#### 왜 이 알고리즘?
N명이 a,b 거듭 번호 줄어드는지 추축해서

#### 시간 복잡도
O(logN)
n은 참가한 인원수, 같은 번호 될떄까지 2로 나눔

#### 공간 복잡도
O(1) 정수형 변수 하나만 사용해서 카운트

#### 엣지 케이스
1라우에서 바로 만남
최종 라운드에서 만남
a>b 인 경우: 대소비교x != 써서 괜찮
#### 새로 배운 것
토너먼트에서 다음 라운드 번호는(x+1) //2
홀짝, 앞의자리를 추가해줘야 나눴을때 정수부가 똑같기 때문에 1추가

#### 문제 정리
두 곳에서 


```python
def solution(n,a,b):
    #대회 참가 n명, 서로 서로 홀+짝 토너먼트
    # 이긴 사람 다음 라운드, 번호 1~n/2번

    #A번을 처음 라운드에서 할당받았는데, B번 가진 참가자랑
    #몇번쨰 라운드에서 만나냐?
    count = 0
    while (a//2 != b//2):
        a = (a+1)//2
        b = (b+1)//2
        count+=1

    return count
```
