#### 왜 이 알고리즘?

#### 시간 복잡도
O(2^N x N)
노드의 수 N
각 상태는 방문 가능한 노드의 집합과 현재 위치의 조합으로 결정
(2^N) = 방문 가능한 노드의 집합

#### 공간 복잡도
O(2^N  * N)
BFS큐에 쌓이는 개수 (2^N)
각 상태마다 set 보관하므로 최대 O(N)개 사용

#### 엣지 케이스
1. 아무런 경우도 돌아올 수 없는 경우 (중복 순환 가능)
2. 루트 직후 바로 늑대만 등장하는 경우 바로 탐색이 종료라 max = 1
3. 자식이 여러개인 경우 


#### 새로 배운 것
1. 현재 상태에서 갈 수 있는 다음 노드들의 집합을 visitied에 set()으로 넣어 관리하기@
2. set.update(iterable), set.add(item)
리스트 전체를 한번에 풀을떄는 update, 단일 원소는 add
3. 차집합 연산, set은 가능 {} 이것으로, 차집합 가능
4. 튜플은 꼭 괄호 (2중으로) 해주기! 

#### 의문


```python
from collections import deque

def solution(info, edges):
    # 0 양 1 늑대 

    # edge 부모 - 자식
    
    #루트에는 양
    #루트 노드에서 각 노드 돌아다니며 양을 모은다
    # 각 노드 방문할때마다, 양과 늑대가 나를 따라온다
    #양 <= 늑대  양 다 잡아먹힘
    #많은 수의 양을 모아서, 루트 노드로 돌아와야 함

    #1. 트리 구축
    def build_tree(info, edges):
        tree = [[]for _ in range(len(info))] #노드의 개수만큼
        #부모 노드에 연결된 자식을 추가
        for edge in edges:
            tree[edge[0]].append(edge[1]) 
            #트리가 여러개 담는 리스트라서, append 쓰는게 좋음
        return tree
            
    #2. 트리 생성
    tree = build_tree(info, edges)

    #3. 최대 양의 수를 저장할 변수
    max_sheep = 0

    #bfs 큐 생성 및 초기 상태 설정
    #(현재 위치, 양의 수, 늑대 수, 방문한 노드 집합)
    #루트노드는 양이기 때문에! 
    q=deque([(0,1,0,set())])

    #bfs 시작
    while q:
        #큐에서 가져오기
        current, sheep_count, wolf_count, visited = q.popleft()
        max_sheep = max(max_sheep,sheep_count)
        #visited = 다음에 방문해 볼 수 있는 노드들?
        visited.update(tree[current])
        #7. 방문한 노드 집합에 현재 노드의 이웃 노드 추
            #앞으로 방문할 자식노드 추가 

        #8. 인접 노드 탐색
        for next_node in visited:
            if info[next_node] == 1:
                if sheep_count != wolf_count+1:
                    q.append((next_node,sheep_count,wolf_count+1,visited- {next_node}))
                #집합 간의 빼기 연산자 -  set에서 뺄 수 있음!
            else: # 양일 경우
                 q.append((next_node,sheep_count+1,wolf_count,visited- {next_node}))

    #양, 늑대 카운트 따로
    #만약 양<= 늑대  -> 하면 방문했다고 표기, 

    #완전탐색으로 양을 최대한 모을 수 있는 선택

    return max_sheep
```
