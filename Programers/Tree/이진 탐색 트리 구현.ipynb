#### 왜 이 알고리즘?
트리: 작으면 왼쪽, 크면 오른쪽이라는 규칙 
일반 리스트에서 데이터 찾으려면 최대 N번, 
트리는 log N

#### 시간 복잡도
O(N^2)
최악의 경우, O(N)
lst aㅏㄴ큼 삽입, serach+lst 길이만큼 탐색하므로
시간복잡도는 O(N^2)

#### 공간 복잡도
반복문 방식 O(1) curr 변수 하나만 갱신하면서 내려가므로 메모리 사용 안함
재귀 O(H) 깊이만큼 스택에 쌓임

#### 엣지 케이스
1. 트리가 비어있는 경우 = None
2. 찾으려는 데이터가 트리에 존재하지 않으면 None 반환해야 함
3. 찾으려는 데이터가 루트면, 
첫 while 문에서 curr.cal !=key 가 거짓이 됨
4. 편향 트리 

#### 새로 배운 것
1. class Node
2. self 데이터를 지속해서 젖장

```python
class Node: #데이터 간의 연결, 데이터 담는 기본 단위
    #data : 저장하고싶은 실제 값
    # next 다음 데이터 주소
    #노드 클래스 생성자, 새로운 객체 초기화 메서드 (생성자)
    def __init__(self, key):
        #init = 객체 만들면 자동으로 호출!!!
        self.left = None
        self.right = None
        self.val = key

#이진탐색 트리 탐색 클래스
class BST:
    def __init__(self):
        self.root = None

    #이진 트리 규칙에 맞는 위치에 새 노드 삽입
    def insert(self, key):
        if not self.root:
            self.root = Node(key)
        #지금 들어온 key으로 root를 만들어라!
        else:
            curr = self.root
            while True:
                #삽입하려는 값이 현재보다 작으면, 왼쪽 노드로 이동
                if key < curr.val:
                    if curr.left: #왼쪽에 변수 있으면
                        curr = curr.left #왼쪽으로 이동
                    else: #왼쪽에 없으면
                        curr.left = Node(key)
                        #클래스 통해서 새로운 객체를 만들기
                        break
                else:
                    #삽입하는 값이 
                    if curr.right:
                        curr = curr.right
                    #오른쪽 노드로 이동
                    else:
                        curr.right = Node(key)
                        break

        def search(self, key):
            curr = self.root
            #노드 존재하고, 찾으련느 값이랑 현재 값이 다르면 반복
            while curr and curr.val != key:
                if key < curr.val:
                    curr = curr.left
                
                else:
                    curr = curr.right
            return curr

        #lst 데이터 활용해서 이진 탐색 트리, search_list 원소 탐색
        def solution(lst, search_lst):
            bst = BST() #이진트리 탐색 트리 클래스

            #리스트 요소로 이진탐색 트리 만들기
            for key in lst:
                bst.insert(key)
            result = []
            #검색 리스트 요소를 검색하고, 결과 리스트에 추가
            for search_val in search_lst:
                if bst.search(search_val):#노드 존재하면 값 반환
                    result.append(True) 
                else:
                    result.append(False) 

                return result




```
