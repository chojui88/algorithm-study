```python
def can_move(board, nx, ny):
    #보드 밖, 발판 없는곳 이동 불가
    if 0<= nx < 5 and 0<= ny < 5 and board[nx][ny] == 1:
        return 1
    return 0

def solution(board, aloc, bloc):
    # 게임 끝날때까지, 양 플레이어 캐릭터 몇번 움직이나? A가 먼저 시작

    # 게임 보드 초기 상태 board
    # A의 위치 aloc
    # B의 위치 bloc

    #만약 움직일 수 있으면
    


    def dfs(board, aloc, bloc,answer):
        #상하좌우
        dx = [-1,1,0,0]
        dy = [0,0,-1,1]

        turn = answer%2
        if turn == 0:
             x, y =aloc
        else:
             x,y = bloc

        for i in range(4):
            nx = x + dx[i]
            ny = y + dy[i]

            # 범위가 넘지 않는지, 움직일 수 있는 1 칸인지
            if can_move(board, nx, ny):

            board[nx,ny] = 0
                
            if answer%2 == 1: #A차례
                aloc = [nx , ny] #이렇게 해도 되나?
            else:
                bloc = [nx , ny] 

            
            #갈 수 있는 곳이하나인데, 그곳이 == 상대방의 위치라면? -> 패비
            # 
            if aloc == bloc:
                        return answer
            
            #움직였으면 x -> 내 순서에서 다음 칸으로 이동할때 0으로 바꿔야 한다
            new_board = board[nx][ny] = 0
            turn +=1
            dfs(new_board, aloc, bloc,turn)

        #발판이 없으면 -> 패배
        return answer

    answer = 0
    dfs(board)
    

    #캐릭터 이동 횟수의 합 
    return answer
```
