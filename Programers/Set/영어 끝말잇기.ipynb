#### 왜 이 알고리즘?
순차적으로 단어 확인하면서 첫번째 인덱스 찾는 것
배열 순회: 앞 단어부터 검사하는 O(N)
in연산으로 중복 찾으면, 시간 많이걸리니까 set 자료형 O(1)

#### 시간 복잡도
단어의 개수를 N, 단어의 평균 길이를 L
시간복잡도 O(N x L)
단어를 순회하고, 각 단어마다 set 조회, 삽입이랑
비교가 일어난다. 문자열 해시 계산에는 단어평균 길이만큼 시간이 소요된다

#### 공간 복잡도
set집합에 최대 N개의 단어를 저장
O(N)

#### 엣지 케이스
1. 첫 단어, 중복 단어 집합에는 들어가야 함
2. 마지막 단어, 
3. 모든 단어가 성공하면 맨 마지막 [0,0] 반환 챙기기
4. 한글자 단어 조건 처리

```python
def solution(n, words):
    used_words = set()
    prev_word = words[0][0]

    for i, word in enumerate(words):
        if prev_word[-1] != word[0] or word in used_words or len(word) <=1:
            return[(i%n)+1, (i//n)+1]

        used_words.add(word)
        prev_word = word #다음 단어
    return [0,0]
```
