#### 왜 이 알고리즘?
해시/ 집합 단원
집합: setdㅔ 저장 = 시간복잡도 O(N x 전화번호길이^2)
정렬: O(N) 인접 비교

#### 시간 복잡도
O(NlonN x L)
sort 알고리즘은 N log N

#### 공간 복잡도
x 제자리 정렬 sort()

#### 엣지 케이스
1. 원소 1개 -> ren(len(phone_book-1))로 인덱스 초과 막음
2. 번호의 길이가 서로 다름 => 사전순 정렬, startswith

#### 새로 배운 것
1. 데이터 범위 너무 클때, 먼저 정렬하면 탐색 범위 NlogN으로 줄일 수 있음
2. startswith() 메소드! 접두사 검사! 불리언 값 얻을 수있음
3. True, False는 문자열에 담지 않도록 조심하기


```python
def solution(phone_book):
    #한 번호가 다른 번호의 접두어, 사전순으로 정렬하고, 
    phone_book.sort()
    #바로 앞 문자열의 접두어인지 확인!
    for i in range(len(phone_book)-1):
            if phone_book[i+1].startswith(phone_book[i]):
                return False

    return True
```
