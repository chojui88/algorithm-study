### 왜 이 알고리즘?
크루스칼 알고리즘 
최소 신장 트리 MST (Minimum Spanning Tree)
최소 비용을 모든 노드 연결
그리디 접근 (적게 드는 방식부터)
사이클 방지  = 유니온 - 파인드 활용!

### 시간 복잡도
시간복잡도 O(다리수 log 다리수 )
len(costs) = 다리수 = E
    정렬: costs.sort O(ElogE)
    유니온 파인드 연산: O(1), O(a(v))
    반복문: 간선 E개를 순회, O(E)

### 공간 복잡도
    O(V + E)
    parent, rank 배열 크기 O(V) 
    정렬 메모리 공간 O(E)
    O(V + E)

```python
def find(parent, i):
    if parent[i] == i:
        #parent는 현재까지 다리를 놓아 연결한 섬의 정보  집합으로 
        return i
    parent[i] = find(parent, parent[i])
    return parent[i]
    #재귀 호출 형태로, 경로 압축

def union(parent, rank, x, y):
    #rank 기반으로 구현! 
    xroot = find(parent, x)
    yroot = find(parent,y)

    if rank[xroot] < rank[yroot]:
        #루트 노드의 랭크가 다르면 랭크가 큰 집합에 작은것을 추가해 합침
        parent[xroot] = yroot
    elif rank[xroot] > rank[yroot]:
        parent[yroot] = xroot
    else:
        parent[yroot] = xroot
        rank[yroot] +=1

def solution(n, costs):
    costs.sort(key = lambda x: x[2])

    parent = [i for i in range(n)]
    rank = [0] * n

    min_cost = 0 #최소 신장트리 총 비용
    edges = 0 #최소 신장트리 포함된 간선의 개수

    for edge in costs: 
        if edge == n-1:
            #섬이 모두 연결되었다
            break

        #현재 간선 두 노드가 속한 루트 찾고
        x = find(parent, edge[0])
        y = find(parent, edge[1])

        if x != y:
            #사이클 없으면 다리를 추가하고, 현재추가한 다리 비용 더함
            union(parent, rank, x, y)
            min_cost += edge[2]
            edges += 1 #간선 개수 증가

    return min_cost
```
