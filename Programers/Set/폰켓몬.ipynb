#### 왜 이 알고리즘?
가질수 있는 포켓몬의 종류가 중복되면 1개로 처리되므로 set집합 사용
Use case 접근법
중복 제거한 포켓몬 수가 충분히 많으면, n/2가 답이고
그렇지 않으면 그 종류의 수가 답이기 떄문에

#### 시간 복잡도
nums 리스트의 길이 N
set(nums) 생성하는 O(N)

#### 공간 복잡도
set_nums의 경우 최액은 O(N)

#### 엣지 케이스
1. 모든 포켓몬이 같은 종류면
len(set_nums)는 1이 되므로 1조건에 걸려 1 반한
2. 서로 다른 종류
if문에서 걸림 
3. 경계값에 걸리는 경우

#### 새로 배운 것
1. 조건문 비교는 // 연산자가 안정성 측면에서 더 좋음

```python
def solution(nums):
    #n 마리 포켓몬 종류 번호 
    #홍박사님의 연구실, 
    # N마리의 포켓몬 중이서 n/2 마리 가져가ㅗㄷ 좋음

    #종류에 따라 다른 번호
    #같은 종류는 같은 번호
    #가질 수 있는 포켓몬의 종류 = set 사용!
    #최대한 다양한 종류의 포켓몬! n/2마리 선택
    
    n = len(nums)
    set_nums = set(nums)
    if len(set_nums) >= n /2:
        return n//2
    else: 
        return len(set_nums)

        
        
    

    
```
