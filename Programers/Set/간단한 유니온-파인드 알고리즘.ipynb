#### 왜 이 알고리즘?
동일 집합 포함 여부 갱신하면서 확인
bfd dfs는 최악의 경우 시간초과

#### 시간 복잡도
N은 operations의 길이!
find, uion 함수의 시간복잡도는 O(1)
parents 초기화할때 크기가 k인 리스트 생성할때, O(k)
이후 operations 연산 N번 수행 O(N)
O(N+K)
#### 공간 복잡도
O(k)
노드 개수에 대해 parent 리스트와 rank_data 리스트를 유지
노드수에 비례!

#### 엣지 케이스
1. k=1인 경우, range(k) 범위 처리
2. 같은 집합인 두 노드를 시도
root1 = root2 조건 분기, 
3. 연쇄 경신 누락
직접 parent[x]만 조회, 최신 루트 노드 못가져오면
find(x,parent)를 통해 검증
4. 재귀 깊이
sys.setrecursionlimit()설정 필수적

#### 새로 배운 것
1. list(range(k))
2. 불리언 리턴 팁 result( == )조건 통새 true, false 넣ㅇ르 수 있음

```python

    #수행할 연산, 
    #union(x,y) 두 집합을 합침 = u
    # find(x) x가 속한 root 원소 찾음  = f ,x, y루트 노드 같은지 find로 확인 

    # 유니온 -> 파인드 순서

    #0이상 k-1 이하의 정수 , k까지 파이썬 돌려야
import sys
sys.setrecursionlimit(100000)

def find(x, parents):
    if parents[x] == x:
        return x
    parents[x] = find(parents[x], parents)
    return parents[x]

def uion_set(x,y,parent,rank_data):
    root1 = find(x, parent)
    root2 = find(y, parent)

    if root1 != root2:
        if rank_data[root1] < rank_data[root2]:
            parent[root1] = root2
        elif rank_data[root1] > rank_data[root2]:
            parent[root2] = root1
        else:
            parent[root1] = root2
            rank_data[root2] +=1

def solution(k, operations):
    parent = [i for i in range(k)]
    rank_data = [0] * k #트리 높이!

    results = []
    #k는 노드의 개수
    for i in range(len(operations)):
        if operations[i][0] == 'u':
            uion_set(operations[i][1],operations[i][2],parent, rank_data)  
        else: 
            x = find(operations[i][1], parent)
            y = find(operations[i][2], parent)
            if x==y:
                results.append('True')
            else: 
                results.append('False')
            
            
    return results
```
