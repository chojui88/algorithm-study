```python
#첫번쨰 도전
# 문제점: 체육복 빌려주는 코드가 없음
def solution(n,lost,reverse):

    #앞뒤 사람들에게만 줄 수있음
    student = { }
    #1. 전체 리스트 만들어서 1로 넣고
    
    for i in range(1,n+1):
        student[i] = 1
        
    #2. 여기서 reverse는 +1
    for i in reverse:
        student[i] +=1
    
    #3. lost는 -1
    for i in lost:
        student[i] -= 1
        
    #4. 값이 1 이상인 키의 count 값 출력
    count = 0
    for i in range(1,n+1):
        if student[i] >= 0:
            count+=1
```

```python
def solution(n,lost,reverse):

    all_student = {i: 1 for i in range(1,n+1)}

    #중복 제거
    lost = list(set(lost))
    reverse = list(set(reverse))

    for r in reverse:
        if r in all_student:
            all_student[r] +=1

    for l in lost:
            if l in all_student:
                all_student[l] -=1 

    for i in range(1,n+1): 
        if all_student[i] == 2:
         
            if i == 1:
                if all_student[2] == 0:
                    all_student[1] = 1  
                    all_student[2] = 1

            elif i == n:
                if all_student[n-1] == 0:
                    all_student[n] = 1  
                    all_student[n-1] = 1

            else: 
                if all_student[i-1] == 0:
                    all_student[i] = 1
                    all_student[i-1] = 1
                    
                
                elif all_student[i+1] == 0:
                    all_student[i] = 1
                    all_student[i+1] = 1

    answer = sum(1 for key in all_student if all_student[key] > 0)
    return answer
```
