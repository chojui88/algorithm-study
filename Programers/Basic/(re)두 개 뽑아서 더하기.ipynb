#기존 코드
def solution(numbers):
    arr = []

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] not in arr:
                arr.append(numbers[i] + numbers[j])

    arr.sort()

    return arr

```python
#스스로 풀은 코드
def solution(numbers):
    #정수 배열
    #서로 다른 인덱스 두개의 숫자 뽑아서
    #더할 수 있는 모든 수!!!
    #오름차순으로 배열 리턴!
    
    #O(n^2) = 10^4 가능! 
    answer = []
    n_plus = {}
    n = len(numbers)
    for i in range(n):
        for j in range(n):
            if i == j:
                continue
            n_plus[numbers[i]+numbers[j]] = n_plus.get(i+j,0) +1
    
    #2. 해쉬 맵에다 결과를 넣기
    #3. key값 배열에 담아서 리턴
    answer = list(n_plus.keys())
    answer.sort()
    return answer
```
