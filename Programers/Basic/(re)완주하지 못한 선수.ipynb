```python
#### 질문
1. 문자열 담긴 배열도 not in 으로 찾을 수 잇나?
2. 문자열이 담긴 리스트에서 해당 문자열을 지울 수 있는 함수도 있나?
3. 리스트 결과값을 문자열로 바꿔서 출력하는 함수는?
-> "".join(리스트 이름)
```

```python
#기존
def solution(participant, completion):
    p_dict = {}

    # 1. 참가자 카운트 (= 연산자 사용)
    for name in participant:
        p_dict[name] = p_dict.get(name, 0) + 1

    # 2. 완주자 차감
    for name in completion:
        p_dict[name] -= 1

    # 3. 콜론(:) 잊지 말고 붙여주기
    for name, count in p_dict.items():
        if count == 1:
            return name
```
