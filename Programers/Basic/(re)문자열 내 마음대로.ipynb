## 질문
1. 첫번째 정렬 기준 정하고 두번째 정렬 기준 지정하려면 식을 어떻게 써야해?
-> 기준들을 : 뒤에 기준을 () 괄호로 묶어서 콤마로 구분
answer = sorted(strings, key = lambda x : (x[n],x))

```python
#기존의 코드
def solution(strings, n):

    # 문자열로 구성된 리스트, 정수 n
    # string 의 인덱스 n글자를 기준으로 오름차순 정렬!

    #만약 인덱스 문자가 여럿이면, 사전순으로 즉 0번 인덱스부터 사전순으로 오름차순! 
    strings.sort(key = lambda x: (x[n],x))

    return strings
```
