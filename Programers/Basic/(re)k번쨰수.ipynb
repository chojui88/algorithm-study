###질문
1. commands가 [i,j,k] 인데 for i,j,k in zip(commands): 이 문법이 맞나
-> 틀림!! 
zip 안쓰고 그냥
 in commands 쓰면 된다!  (파이썬 언페킹)

 ###실수
 1. k 인덱싱 주의!!

```python
#원래 코드
def solution(array,commands):
    #arr 배열 i부터 j까지 고르고, 정렬했을때 k번째에 있는 수
    #commands는 i,j,k를 가진 수! 

    answer = []
    for i,j,k in commands:
        new_arr = array[i-1:j]
        new_arr.sort()
        answer.append(new_arr[k-1])
    
    return answer
```
