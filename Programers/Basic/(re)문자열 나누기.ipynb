Q. 이거 문자열을 슬라이싱해서 pop하고 싶은데 while 문 안에 
while s:
    s = [s[num+1]:]
이렇게 해서 s 를 슬라이싱해서 복사해도 써도 되나? 이거 shallow copy잖아

A. 가능하다 대신 
s = s[num+1:] -> 이게 맞는 표현!

파이썬은 불변 객체! => 기존 문자열읊 직접 수정할 수 없다

```python
def solution(s):
    #첫 글자 x가 아닌 다른 글자들이 나온 횟수 세기
    # 1. 첫 글자 x 저장하고
    
    # 2. x와 동일한 글자 same 변수, 다른 글자 diff 변수에 수를 각각 카운트
    count = 0 #숫자를 셌어 
    while s:  #1) banana 
        same, diff = 0, 0 
        x = s[0] #x = b
        
        for num in range(len(s)):  #num == 5
 
                
            if s[num] == x : #지금 s[1] == 'a'
                same +=1
                
            else: #다르다면?
                diff +=1
            
            if same != 0 and same == diff:
                s = s[num:]
                count += 1
                break
          
        
    #3. 두 횟수가 같아지는 순간 지금 카운팅한 문자 까지를 슬라이싱 해서 pop 하고, 이를 count+=1 함
    
    #4. 이 과정을 끝까지 반복해서 최종 count 개수 반환
    return count
```

```python
def solution(s):
    count = 0

    while s:
        same, diff = 0, 0
        x = s[0]

        for num in range(len(s)):
            if s[num] == x:
                same += 1
            else:
                diff += 1

            if same == diff:
                s = s[num+1:]
                count += 1
                break
        else:
            count += 1
            break

    return count
```
