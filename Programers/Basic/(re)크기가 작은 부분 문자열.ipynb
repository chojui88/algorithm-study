```python
##질문
1. np가 [] list형태인데, 이대로 리스트에 넣으면 [[]] 이렇게 리스트 안에 리스트잖아?
이 리스트 안에 리스트 숫자들을 set으로 해서 숫자 형식으로 count함수를 적용시킬 수 있어?

-> 파이썬에서 [] 이게 항상 리스트만을 의미하지는 않는다! 
슬라이싱을 하는 대상 t 가 문자열이면, 문자열이 저장되는 거고
리스트를 슬라이싱 하면 리스트가 저장된다

2. set으로 정의된 answer의 원소 개수 세는 함수 count 아니야? 라이브러리 블러와야하나?
count는 함수로 .count(x) 이렇게 써서, 특정값이 존재! 하는지 세는 함수! 
이럴땐 len() 써야함
```

```python
def solution(t, p):
    #p의 길이 10^5
    #숫자로 이루어진 문자열
    #p의 길이를 가진 부분문자열 중에 , p보다 수가 작거나 같은 횟수 
    count = 0

    #2. for문 돌려서 슬라이싱으로 문자열 하나하나 값 비교하면 되겠네 
    np = []
    answer = []
    for i in range(len(t)-len(p)): #i는 0부터 7까지 반복 -> t[0:2] ~ t[2:5]
    
        np = t[i:i+len(p)]
        
    #3. 그 중에서 값이 p보다 작은 수면 count +=1 하기
        if np <= p:
            answer.append(np.join(""))
            answer = set(answer)
    
    return count(answer)
```
