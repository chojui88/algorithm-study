##질문
2차원 배열을 그냥 
sizes.sort() 하면 그 배열 자채가 정렬이 되는건가?
2차원 배열 속의 [a,b] 안에 있는 그 a,b를 정렬하려면 어떻게 해?

sizes.sort()하면 각 내부 배열의 첫번째 원소를 기준으로 오름차순
-> 반복문 사용해서, 
for size in sizes:
    size.sort()
그 배열 안에 있는 배열을 정렬해야 한다! 

```python
def solution(sizes):
    wide = []
    high = []
    for i in sizes:
        n = 0
        w,h = i
        if w<h:
            n = h
            h = w
            w = n
        wide.append(w)
        high.append(h)
    return max(wide) * max(high)
```
