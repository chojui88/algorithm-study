3. 딕셔너리에서 키 v가 몇번 등장했는지 세는 코드
찾으려는 키
numbers.get(v,0) 못찾으면 반환값
그렇게 키의 valueㄱ ㅏㅂㅅ을 가져운 것에 +1을 해야한다! 

```python
#원래 코드
def solution(nums):
    #n 마리 포켓몬 종류 번호 
    #홍박사님의 연구실, 
    # N마리의 포켓몬 중이서 n/2 마리 가져가ㅗㄷ 좋음

    #종류에 따라 다른 번호
    #같은 종류는 같은 번호
    #가질 수 있는 포켓몬의 종류 = set 사용!
    #최대한 다양한 종류의 포켓몬! n/2마리 선택
    
    n = len(nums)
    set_nums = set(nums)
    if len(set_nums) >= n /2:
        return n//2
    else: 
        return len(set_nums)
```
