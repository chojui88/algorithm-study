int 형 숫자의 길이, 자릿수를 세는 코드를 구현하는 함수나 구문은?
-> len(str(num))

```python
def solution(x):
    #x가 하샤드 수려면, x의 자릿수의 합으로 x가 나눠져야 한다 
    
    s_num = 0
    #1. 입력받는 수의 각각 자릿수의 수를 추출한다
    for i in range(len(x)):
        s_num += (x % 10**i)  #1의 자리 수 ~ 10의자리 수
    #2. 각 수를 더한다 -> s_num 저장
    
    #3. 더한 수를 x로 나눴을때 나머지가 0인지 검사한다
    if x%s_num == 0:
        return true
    else:
        return false
    
    #4. 나머지가 0이라면 true 반환! 
    
    #5. 아니라면 false 반환
```
