-> sorted()와 lambda를 사용하면 가능하다
sorted()로 fail_dic를 먼저 정렬을 하고 그 다음에 키로 정렬하면 된다
result = sorted(fail_dic, key = lambda x: (-fail_dic[x], x))
여기서 sorted자체가 새로운 리스트를 반환하기 때문에 result = []를 미리 선언할 필요 없다

```python
def solution(N, stages):
    answer = []
    fail =[0] * (N + 1)

    #스테이지에 있는 사람들의 수 하나하나를 fail의 그 인덱스에 값으로 1명, 2명 각각 1111이런식으로 채우고싶어
    fail[0] = len(stages)
    for i in stages:
        for j in range(1,i):
            fail[j]+=1

    #스테이지에 있는 값 하나하나에 대한 실패율을 구하고 이를 정렬하고싶어
    for i in range(1,N+1):
        if fail[i-1] == 0:
            fail_rate = 0
        else:
            fail_rate = (fail[i-1]-fail[i]) / fail[i-1]

        answer.append((i,fail_rate))

    
    answer.sort(key=lambda x: -x[1])
    return [x[0] for x in answer]
```

```python
#다음으로 수정한 코드

def solution(N, stages):
    #N: 전체 스테이지 수
    #stages: 멈춰있는 스테이지 번호 , n+1 까지의 번호가 있음 (마지막 클리어 n+1)
    
    stage_dic = {}
    for i in stages:
        stage_dic[i] = stage_dic.get(i,0) + 1
        #stage_dict 속에는 stages를 깬 1번부터 n+1까지의 수가 담겨있다 n은 도전중인 스테이이므로 i-1를 해야, 깐 스테이지가 된다
        #n+1 = 모든 스테이지를 클리어한 사람 수 
        #1이라면, 시작한 사람! 
        
    
    #1. 실패율 구하기
    #스테이지에 도달했으니 아직 클리어하지 못한 플레이어 수 / 도달한 플레이어 수
    # 즉 이전 단계를 꺴지만 현재단계는 꺠지 못한사람 / 이전 단계를 깬 사람
    
     # 이전 단계를 깬지만 현재는 깨지 못한 사람의 합 = 지금 이전단계에 머물러 있는 사람 
    fail_dic = {}
    
    
    for i in range(1,N+2):
        fail = 0
        # 이전 단계를 깬 사람의 합
        sum = 0
        #i는 1부터 n+1까지인데
        # 각각 이전에 깬 스테이지이다. 1이면 1까지 깬것! 근데 아에 안깰수도 있긴 함 stages는 1부터 시작하니까? 그래서 0부터 해야
        
        #stage_dic 안에, 아에 이전단계 깬 단계가 있음 그래서 0~N까지만 있는 것임! 
        for j in range(i,N+2):
                sum += stage_dic.get(j,0)
                #여기서 i는 그 스테이지에 도달한 사람 수
                #j도 그 스테이지를 도달한 사람수
                #하지만 우리는 이전단계를 꺤 사람을 저장해야 한다 그래서 j인덱스에 있는 사람부터 끝까지 더하면 됨! 
                #여기서 i가 1이라면, j가 1일때도 시작한 사람
                
        #fail_dic 안에는, 해당 key값은 현재 위치한 단계
    #그리고 values는 현재 위치한 단계를 갠 사람 수의 합! 
        if sum == 0 :
            fail_dic[i] = 0
            #그런데 만약 i인 곳에 합이없으면 그 이후의 값이 없다는 것이니까 그 이전단계를 깬 사람이 없다는 거니까 i의 단계 이전단계를 깬 사람이 없다는 듯,
            
        #stage_dic의 키값은 현재 스테이지를 도달한 사람
        #stage_dic이 만약에 n+1이라면, 마지막 스테이지를 
        fail = stage_dic[i-1] / sum
        fail_dic[i] = fail
    

    #2. 실패율이 높은 스테이지부터 내림차순으로 스테이지 번호가 담겨있는 배열 return
    #2-1. 실패율이 같다면 작은 번호의 스테이지가 먼저 오도록, 실패율 정렬 => stage number 정렬
    fail_sort_stages = sorted(fail_dic, key = lambda x: (-fail_dic[x], x) )
    
    retur
    fail_sort_stages
```
