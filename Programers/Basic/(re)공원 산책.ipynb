```python
#include <string>
#include <vector>

using namespace std;

vector<int> solution(vector<string> park, vector<string> routes) {
    vector<int> answer;

    int x, y;   // x = 가로, y = 세로

    // 시작 위치 찾기
    for (int i = 0; i < park.size(); i++) {
        for (int j = 0; j < park[i].size(); j++) {
            if (park[i][j] == 'S') {
                x = j;
                y = i;
            }
        }
    }

    int dx = 0;
    int dy = 0;

    // 명령 수행
    for (int i = 0; i < routes.size(); i++) {

        char dir = routes[i][0];
        int dist = routes[i][2] - '0';

        dx = 0;
        dy = 0;

        if (dir == 'N') {
            dx = 0;
            dy = -1;
        }
        else if (dir == 'S') {
            dx = 0;
            dy = 1;
        }
        else if (dir == 'W') {
            dx = -1;
            dy = 0;
        }
        else if (dir == 'E') {
            dx = 1;
            dy = 0;
        }

        bool canMove = true;

        for (int k = 1; k <= dist; k++) {

            int nx = x + dx * k;
            int ny = y + dy * k;

            // 공원 밖인지 확인
            if (nx < 0 || nx >= park[0].size() ||
                ny < 0 || ny >= park.size()) {
                canMove = false;
                break;
            }

            // 장애물 확인
            if (park[ny][nx] == 'X') {
                canMove = false;
                break;
            }
        }

        // 이동 가능하면 이동
        if (canMove) {
            x += dx * dist;
            y += dy * dist;
        }
    }

    answer.push_back(y);   // 세로
    answer.push_back(x);   // 가로

    return answer;
}
```
