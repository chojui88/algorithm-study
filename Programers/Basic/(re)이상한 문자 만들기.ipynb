##질문
1. 공백에 upper, lower적용해도 공백으로 나오지?
-> 응 그대로 공백이 반혼 돼

2. i로 해서 문자열에s[i] 이렇게 하면 공백은 인식이 안되나? 

```python
def solution(s):
    
    #s: 한 개 이상의 단어, 하나 이상의 공백문자
    answer = []
    #1. 각 문자열을 분리해서
    for i in range(len(s)):
    #2. 0부터 시작해 짝수번째 알파벳은 대문자 %2 == 0 
        if i%2 == 0:
            answer+(s[i].upper())
        else:
            answer+(s[i].lower())
    #3. 홀수번쨰는 소문자로 새로운 문자열에 저장한다
    
    return answer
```
