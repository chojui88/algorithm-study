```python
def solution(arr1, arr2):
    rows1 = len(arr1)
    cols1 = len(arr1[0])
    cols2 = len(arr2[0])

    answer = [[0] * cols2 for _ in range(rows1)]

    for j in range(rows1):
        for k in range(cols2):
            for i in range(cols1):
                answer[j][k]+=arr1[j][i] * arr2[i][k]  
            
    return answer
```
