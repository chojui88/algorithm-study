#### 왜 이 알고리즘?
n x n 정사각형 격자
숫자는 그 칸의 높이!
모든 칸 방문하는 탐험 하려고
상하좌우
사다리 설치 , (height보다 높은 높이차) 두 격차 칸의 높이 차 만큼 비용이 든다

bfs
#### 시간 복잡도
land 한 변의 길이 N 
각 지점 방문하는 시간복잡도는 O(N^2)

우선순위큐 너비 우선 탐색
O(N^2 log(N^2))
전체 칸수 x 힙의 연산 비용, 원소 하나 넣거나 꺼내는데 log

#### 공간 복잡도

#### 엣지 케이스
1. 현재칸과 이동하려는 높이 차가 height 이하여야 한다
2. 상하좌우 이동

#### 새로 배운 것

#### 질문

```python

from heapq import heappush, heappop
#힙은 비용이 적은 원소부터 꺼냄

def solution(land, height):
    #land: 각 칸의 높이
    #height : 사다리 설치 기준

    answer = 0
    n = len(land)

    #상하좌우로 옮겨보고
    dx = [-1,1,0,0]
    dy = [0,0,-1,1]
    visited = [[False] * n for _ in range(n) ]

    q = []
    heappush(q, [0,0,0])

    while q: 
        cost, x, y = heappop(q)
        if not visited[x][y]:
            visited[x][y] = True

            #현재가지 비용을 합산
            answer += cost

            for i in range(4):
                nx = x + dx[i]
                ny = y + dy[i]

                #범위를 넘지 않는가?
                if 0<= nx < n and 0<= ny < n:

                    if abs(land[nx][ny]-land[x][y]) > height:
                        #사다리 설치
                        # 다음 칸을 이동할때 드는 비용
                        new_cost =abs(land[nx][ny]-land[x][y])
                    else: 
                        new_cost = 0
                    heappush(q,[new_cost, nx,ny])

                
    #최대한 적은 사다리 설치 비용 반환
    return answer
```
