### 왜 이 알고리즘?
튜플: 셀 수 있는, 순서있는
    중복된 원소 가능
    원소 순서가 다르면 다른 튜플
    중복되는 원소가 없는 튜플에서, 집합 기호로 표현 가능 {} -> 튜플로만들 수 있는 개수의 조합하
    단 집합은 순서 바뀌어도 상관없음

    튜플 집합 S 
    s가 표현하는 튜플을 배열에 담아 return 하도록

### 시간 복잡도
s는 10^6승 이하니까, 시간복집도는 N^2 까지 가능
s, 튜플 집합의 길이를 N이라고 하고
s를 split()한 결결과로 얻은 리스트의 길이를 M이라고 하면
처음에 split()의 시간복잡도는 O(N) 이고
이를 정렬하기 위해서 필요한 시간복잡도는 O(MlogM)

```python
#원소 개수가 적은 집합을 오름차순으로 정러란다음, 이전 집합과 비교해서 중복된 원소 제거하는 방식

def solution(s):
    #문자얄 s를 파싱해서, 리스트로 변환
    s = s[2:-2].split("},{")
    s = sorted(s, key = len)

    answer = []
    #각 원소를 순회하면서 차이 나는 부분을 구하기
    for group in s:
        for i in group.split(","):
            if int(i) not in answer: 
                answer.append(int(i))

    return answer
```
