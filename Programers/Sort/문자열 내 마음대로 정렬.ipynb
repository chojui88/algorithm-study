### 새로 배운 것
1. 원하는 인덱스를 기준으로 배열하고싶다? key = lambda!!
그 다음 주건은?  lambda x : 다음에 괄호로 묶기!!


```python
def solution(strings, n):

    # 문자열로 구성된 리스트, 정수 n
    # string 의 인덱스 n글자를 기준으로 오름차순 정렬!

    #만약 인덱스 문자가 여럿이면, 사전순으로 즉 0번 인덱스부터 사전순으로 오름차순! 
    strings.sort(key = lambda x: (x[n],x))

    return strings
```
