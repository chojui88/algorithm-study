```python
def solution(dirs):
    graph = [0,0]#지금 현위치 마지막 0 은 깃발!
    visited = set()
    count = 0

    for i in dirs:

        before = (graph[0], graph[1])

        if i == 'U':
            graph[1]+=1
        elif i == 'D':
            graph[1] -=1
        elif i =='R':
            graph[0] +=1
        elif i =='L':
            graph[0] -=1

        after = (graph[0], graph[1])

    #그래프를 한번 거쳐서
        if graph[0] < -5 or graph[0] > 5 or graph[1] < -5 or graph[1] > 5:
            graph[0], graph[1] = before
            continue
        path = (before, after)
        reverse_path = (after, before)

        if path not in visited and reverse_path not in visited:
            visited.append(path)
            count+=1

    return count
```

```python
def is_valid_move(nx,ny):
    return 0<= nx < 11 and 0 <= ny < 11

def update_location(x,y,dir):
    if dir == 'U':
        nx, ny = x, y+1


    return nx, ny

def solution(dirs):
    x,y = 5,5
    ans = set() #지나간길 저장!!
    for dir in dirs: #명령어
        nx, ny = update_location(x,y,dir)
        if not is_valid_move(nx,ny):
            continue
        ans.add((x,y,nx,ny))
        ans.add((y,x,ny,nx))
        x,y = nx,ny
    return len(ans)/2
```
