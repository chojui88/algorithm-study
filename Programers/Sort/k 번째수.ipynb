### 문제
1. 이거 인덱스 안터질려면 어떻게 해야하지? 

```python
def solution(array,commands):
    #arr 배열 i부터 j까지 고르고, 정렬했을때 k번째에 있는 수
    #commands는 i,j,k를 가진 수! 

    answer = []
    for i,j,k in commands:
        new_arr = array[i:j+1]
        new_arr.sort()
        answer.append(new_arr[k])
    
    return answer
```

```python
def solution(array,commands):
    #arr 배열 i부터 j까지 고르고, 정렬했을때 k번째에 있는 수
    #commands는 i,j,k를 가진 수! 

    answer = []
    for i,j,k in commands:
        new_arr = array[i-1:j]
        new_arr.sort()
        answer.append(new_arr[k-1])
    
    return answer
```
