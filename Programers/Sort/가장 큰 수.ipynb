```python
import functools
#

def compare(a,b):
    #numbers <= 10^5 = 시간복집도는 NlogN 이하로 풀어야 한다!

    #1. 각 수를 문자열로 바꾸고, 조합하여 더 큰 수를 반환
    t1 = str(a) + str(b)
    t2 = str(b) + str(a)

    return (int(t1) > int(t2)) - (int(t1) < int(t2))
    # 두 숫자를 비교해서, 1, 0, -1 중 하나 반환
    # 문자열을 정수로 변환 int()
    
def solution(numbers):
    # 2. 내림차순으로 정렬
    sorted_nums = sorted(
        numbers,  #numbers를 새로 정렬한 리스트 반환
        key = functools.cmp_to_key(lambda a, b: compare(a,b)), 
        #어떤 기준으로 정렬? 비교하는 함수 기준으로! compare(a,b)
        #우리가 만든 함수를 기준으로 쓰는 것!!!!! 
        #key가 정렬 기준을 뽑아내는 건데, compare()은 원소 두개를 비교하므로
        #compare 함수를 sorted() 함수 안에서 사용할 수 있도록 변환해주는 함수
        reverse = True
    )

    #각 정렬된 수를 문자열로 이어붙여서, 
    answer = "".join(str(x) for x in sorted_nums)
    return "0" if int(answer) == 0 else answer
```

```python
# 이 해결방법의 문제점! 각 자리수가 30, 3 인 경우는 30이 먼저 앞에 오면 안된다, 

from collections import deque

def solution(numbers):
    q = deque()
    # 양의 정수, 정수 이어붙여 만들수 있는 가장 큰 수
    answer = []
    for i in numbers:
        answer.append(list(str(i)))

    #[[1,0],[1,2]]
    answer.sort(reverse = True)

    result = []
    q = deque(answer)
    while q:
        result.append( "".join(q[0])) #가장 큰 수 
        q.popleft()

    return "".join(result)
    #가장 큰 수 

    
```
