```python
def solution(N, stages):
    answer = []
    fail =[0] * (N + 1)

    fail[0] = len(stages)
    #스테이지에 있는 사람들의 수 하나하나를 fail의 그 인덱스에 값으로 1명, 2명 각각 1111이런식으로 채우고싶어
    for i in stages:
        for j in range(1,i):
            fail[j]+=1

    #스테이지에 있는 값 하나하나에 대한 실패율을 구하고 이를 정렬하고싶어
    for i in range(1,N+1):
        if fail[i-1] == 0:
            fail_rate = 0
        else:
            fail_rate = (fail[i-1]-fail[i]) / fail[i-1]

        answer.append((i,fail_rate))

    
    answer.sort(key=lambda x: -x[1], reverse=True)
    return [x[0] for x in answer]
```

```python
def solution(N, stages):
    challenger = [0] * (N+2)
    for stage in stages:
        challenger[stage] += 1

    fails = {}
    total = len(stages) 

    for i in range(1, N+1):
        if challenger[i] == 0:
            fails[i] == 0
        else:
            fails[i] = challenger[i] / total
            total -= challenger[i]

    result = sorted(fails, key = lambda x : fails[x], reverse=True)
```
