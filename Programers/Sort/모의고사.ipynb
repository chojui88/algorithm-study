```python
def solution(answer):
    arr1 = []
    arr2 = []
    arr3 = []

    pattern2 = [2, 1, 2, 3, 2, 4, 2, 5]
    pattern3 = [3, 3, 1, 1, 2, 2, 4, 4, 5, 5]

    # 각 사람이 찍는 답안 만들기
    for i in range(len(answer)):
        arr1.append(i % 5 + 1)
        arr2.append(pattern2[i % 8])
        arr3.append(pattern3[i % 10])

    # 점수 계산
    count1 = 0
    count2 = 0
    count3 = 0

    for i in range(len(answer)):
        if answer[i] == arr1[i]:
            count1 += 1

        if answer[i] == arr2[i]:
            count2 += 1

        if answer[i] == arr3[i]:
            count3 += 1

    scores = [count1, count2, count3]

    # 가장 높은 점수
    max_score = max(scores)

    result = []

    for i in range(3):
        if scores[i] == max_score:
            result.append(i + 1)

    return result
```
