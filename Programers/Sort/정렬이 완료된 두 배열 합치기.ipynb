```python
def solution(arr1, arr2):
    #병합 정렬하는 함수!
    #각각 오름차순
    merged = [] #정렬된 배열을 저장할 최종 리스트
    i, j = 0,0 #두 배열의 인덱스!

    #두 배열 순회하면서 정렬 배열 생성
    for i, j in zip(arr1, arr2):
        if i>j :
            merged.append(i)
            merged.append(j)
        else: 
            merged.append(j)
            merged.append(i)
            
    #arr1, arr2 중 남아있는 원소 정렬 배열 뒤에 추가

    return #정렬 결과 문자열 merged
```

```python
def solution(arr1, arr2):
    #병합 정렬하는 함수!
    #각각 오름차순
    merged = [] #정렬된 배열을 저장할 최종 리스트
    i, j = 0,0 #두 배열의 인덱스!

    while i< len(arr1) and j < len(arr2):
        #작은것 하나를 merge에 넣음
        if arr1[i] < arr2[j]:
            merged.append(arr1[i])
            i+=1
        else: 
            merged.append(arr2[j])
            j +=1
        #그쪽 포인터 +1

        #다시 비교

        #한쪽 배열 끝날때까지 반복
    while i < len(arr1):
        merged.append(arr1[i])

    while j < len(arr2):
        merged.append(arr2[j])

        #남은 배열 그대로 붙임

    return merged
```
