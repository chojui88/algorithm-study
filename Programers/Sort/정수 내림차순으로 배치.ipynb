다시 문자열로 조합하는것은 logN이 필요하다

### 새로 배운 것
1. str(정수) = 정수를 문자열로 바꿔주는 함수
2. list() = 문자열을 리스트 문자 하나하나 담긴 리스트로 바꿔주는 함수 
=> str(134556) 이걸 ['1','2','3'] 
3. 그리고 정수로 다시 반환할때는 "".join(문자열) 해주면 됨!!!  이거 int로 감싸서
int("".join(문자열))
4. sort() 함수는 리스트 자체를 정렬
sorted() 함수는 복사본!!!!!!!

```python
def solution(n):
    #n정수의 각 자리수를 큰것부터 작은순으로 정렬한 새로운 정수 리턴 
    return n.sort(reverse = True)
```

```python
#각 자리 숫자를 분리해서 정렬해야 함
def solution(n):
    digits = list(str(n)) #정수 n을 분자열로 변환하고 각 지리수를 리스트로 저장
    #list() 이게 str(134556) 이걸 ['1','2','3'] 이렇게 나눠줌
    digits.sort(reverse=True)
    answer = int("".join(digits)) 
    

    return answer
```
