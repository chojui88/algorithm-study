```python
def solution(sequence, k):
    lst = []
    # j가 i부터 끝까지 들어갈 수 있도록 range 조정
    for i in range(len(sequence)):
        for j in range(i, len(sequence)):
            # sequence[i:j+1]의 합이 k인지 확인
            if sum(sequence[i:j+1]) == k:
                lst.append([i, j])
                
    # 길이(j - i)가 가장 짧은 구간 찾기
    # 같을 경우 앞선 인덱스가 우선되도록 min의 key 지정
    answer = min(lst, key=lambda x: (x[1] - x[0]))
    return answer
```

```python
#시간복잡도 O(N2), sum 때문에
def solution(sequence, k):
    #수열, 
    left = 0
    right = 0

    lst = []

    #인덱스 사이 원소의 합이 k인, 길이가 가장 적은 부분수열
    #왼쪽 포인터부터 오른쪽 포인터 사이에 수열 값의 sum을 구하거

       #값이 k보다 작으면, right를 오른쪽으로
        #값이 k 크면, left를 오른쪽으로 
    while right < len(sequence):
        if sum(sequence[left:right+1]) <k :
            right+=1
        elif sum(sequence[left:right+1]) >k:
            left+=1
            # 값을 찾으면 left, right값 하나씩 증가시키기
        else: 
            lst.append([left,right])
            left+=1
            right+=1

    #결과값은 left, right 인덱스 반환!
    return min(lst, key=lambda x : x[1]-x[0])


```

```python
#시간복잡도 O(N), while 루프n, 함수 전진 n

def solution(sequence, k):
    left = 0
    right = 0
    lst = []

    #포인터가 움직일 때 값을 더하고 빼는 방식으로!
    current_sum = sequence[0]

       #값이 k보다 작으면, right를 오른쪽으로
        #값이 k 크면, left를 오른쪽으로 
    while right < len(sequence):
        if current_sum <k :
            right+=1
            #1을 더했을때, 범위를 벗어나면 sum할 수 없음!
            if right < len(sequence):
                #current_sum에 넓힌 값을 추가하기
                current_sum += sequence[right]

        elif current_sum >k:
            #current_sum에 좁힌 값을 먼저! 빼기
            current_sum -= sequence[left]
            left+=1

            # 값을 찾으면 left, right값 하나씩 증가시키기
        else: 
            lst.append([left,right])
            current_sum -= sequence[left]
            left+=1

    #결과값은 left, right 인덱스 반환!
    return min(lst, key=lambda x : x[1]-x[0])
```
