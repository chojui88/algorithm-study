```python
def solution(prices):
    answer = []
    for i in prices:
        count = 0
        for j in range(len(prices)):
            if i < prices[j]:
                continue
            else:
                count+=1
        answer.append(count)

    return answer
```

```python
def solution(prices):
    n = len(prices)
    answer = [0] * n

    stack = [0]
    for i in range(1,n):
        while stack and prices[i] < prices[stack[-1]]:
            # 주식 가격과 스택 위에 이전 주식을 비교하는데, 
            j = stack.pop()
            answer[j] = i-j
        stack.append(j)
    while stack:
        j = stack.pop()
        answer[j] = n - 1 - j
    return answer
```
