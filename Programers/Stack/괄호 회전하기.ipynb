```python
def solution(s):
    stack = []
    count = 0
    n = len(s)
    for i in range(n):
        rotated = s[i:] + s[:i]
        stack = []
        is_valid = True

        #회전된 문자열에서 하나씩 검사
        for char in rotated:
            if char == '(' or char == '{' or char == '[':
                stack.append(char)
            else:
                if stack and stack[-1] == '(' and char == ')':
                    stack.pop()
                elif stack and stack[-1] == '{' and char == '}':
                    stack.pop()
                elif stack and stack[-1] == '[' and char == ']':
                    stack.pop()
                else:
                    is_valid = False
                    break

        if is_valid and not stack:
                count +=1
    return count
```

```python
def solutions(s):
    answer = 0
    n = len(s)
    for i in range(n):
        stack = []
        for j in range(n):
            c = s[(i+j)%n]
            if c == "(" or c == "[" or c == "{":
                stack.append(c)
            else:
                if not stack:
                    break
                if c == ")" and stack[-1] == "(":
                    stack.pop()
                elif c == "]" and stack[-1] == "[":
                    stack.pop()
                elif c == "}" and stack[-1] == "{":
                    stack.pop()
                else:
                    break
        else:
            if not stack:
                answer+=1
    return answer
```
