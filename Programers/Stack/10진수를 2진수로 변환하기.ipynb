```python
def solution(i):
    stack = []
    while (i//2 >0):
            stack.append(i%2)
            i = i//2
    stack.append(i%2)
    stack.reverse()

    return "".join(map(str, stack))
```

```python
def solution(decimal):
    stack = []
    while decimal>0:
        remainder = decimal %2
        stack.append(str(remainder))
        decimal //=2
    binary = ""
    while stack:
        binary += stack.pop()

    return binary
```
