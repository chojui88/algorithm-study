```python
import math

def solution(progress, speeds):
    answer = []
    n = len(progress)

    #각 작업의 배포 가능일을 구한다, 며칠 남았는지
    days_left = [math.ceil((100 - progresses[i]) / speeds[i]) for i in range(n)]

    count = 0 #배포될 작업 수
    max_day = days_left[0] #가장 늦게 배포될 작업의 가능일

    for i in range(n):
        if days_left[i] <= max_day:
            count+=1
        else: 
            answer.append(count)
            count = 1
            max_day = days_left[i]

    answer.append(count)
    return answer
#첫번째 작업일보다 빠른 작업들은 함꼐 
#늦은 작업은 뒤로 계속 넣음

```
