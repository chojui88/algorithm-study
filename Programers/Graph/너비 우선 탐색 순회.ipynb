### 왜 이 알고리즘?

### 시간 복잡도
O(v+E)
인접리스트 변환할때, 모든 간선 확인하므로 E간선 수 O(E)
bfs 탐색할때 , visited 셋에 노드 개수만큼 큐에 담길 수 있다 O(V)
    그리고 각 노드 방문할때, 이웃 간선을 순회하므로 간선 수 + O(E)
O(E) = O(V+E) = O(V+E)

### 공간 복잡도
O(v+E)
인접 리스트 : 모든 노드v와 연결 정보e를 저장한다 O(v+e)
visited 셋: O(V) 노드 담는다
result 리스트 O(v)

### 엣지 케이스
단일 노드 = queue에 넣어지고 바로 while q:에 의해 탐색 종료
고립된 그래프 = 시작 노드가 연결되어 있지 않다면, 역시 탐색 종료
비연결 마찬가지
방향성/무방향성지 = 무방향이면 adj_list[v].append(u)도 추가되어야 한다! 

### 새로 배운 것
1. defaultdict
2. 전역 변수 사용
3. 자료구조 방문 체크 set()으로 시작복잡도 O(1)
4. dict.get() 간선이 없는 정접에 접근할때,

```python
from collections import deque

def solution(start, graph):
    #출발노드, 도착노드


    q = deque([start])
    visited = set([start])
    answer = [start]

    while q:
        cur = q.popleft()

        for u, v in graph:
            if u == cur and v not in visited:
                visited.add(v)
                q.append(v)
                answer.append(v)

    
    return answer
```

```python
from collections import defaultdict, deque

#전역 변수 선언, 입력받은 그래프를 인접 리스트로 변환!!! 
#
adj_list = defaultdict(list)
#존재하지 않는 키에 접근할때, 자동으로 빈 리스트 생성! 
visited = set()
result = []

def bfs(start):
    # 처음 방문할 노드 큐에 넣고
    queue = deque([start])
    visited.add(start)
    result.append(start)

    while queue:
        node = queue.popleft()

        #인접 이웃 노드에 대해 방문하지 않은 노드 부시
        for neighbor in adj_list.get(node, []):
            #노드별 연결된 이웃 
            if neighbor not in visited:
                queue.append(neighbor)
                visited.add(neighbor)
                result.append(neighbor)
                

def solution(start, graph):
    for u, v in graph:
        adj_list[u].append(v)
    bfs(start)
    return result
```
