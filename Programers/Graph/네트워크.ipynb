### 문제
네트워크 덩어리의 개수 반환하도록 

### 왜 이 알고리즘?
덩어리 문제 = dfs

### 시간 복잡도
dfs를 시도하는 최대 경우의 수는 , 컴퓨터의 수 O(C)
일반적인 DFS는 O+E
모든 노드 방문 visited 때문에 O(N)인데,
각 노드에서 연결되었나 메인 함수에서 한번 더 확인하기 때문에 O(N^2)

### 공간 복잡도
재귀함수의 스택 수 O(C)
visited 배열의 크기

### 엣지 케이스
1. 경계선 넘는 경우
2. 각 컴퓨터 n인 경우 (n-1 까지이다)

### 새로 배운 것


```python
#현재 노드와 연결되고 방문하지 않은 노드들 끝까지 탐색
def dfs(computers, visited, node):
    #현재 노드 방문 처리
    visited[node] = True
    #연결되어 있으며 방문하지 않은 노드라면 (for 문)
    for i in range(len(computers)):
        if computers[node][i] == 1 and not visited[i]:
            #해당 노드르 방문하러 이동
            dfs(computers, visited, i)

#메인 함수 : dfs 전체적으로 돌 수 있게끔 해서, 방문하지 않은 컴퓨터 만날때마다 count 1씩 더함
def solution(n, computers):
    #n 컴퓨터 개수, 연결정보 computers
    
    answer = 0
    visited = [False] * n

    for i in range(n):
        if visited[i] == False:
             dfs(computers, visited, i)
             answer+=1
        #아직 방문하지 않았다면, 해당 노드를 시작으로 깊이 우선 탐색 진행

 #dfs로 연결된 모든 노드 방문하면서 네트워크 개수 증가
    return answer


```
