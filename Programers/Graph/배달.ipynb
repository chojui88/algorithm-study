### 문제
N개의 마을 
1번부터 N버까지 부여 
마을에 양방향 통로가 있고
가중치 있고
1번이 기준!

### 자료구조
힙: 우선순위큐 = 거리가 짧은 노드가 항상 최상단!
꺼내는데 O(1), 새로운거 넣고 정렬하는데 O(logN)
전체 O(ElogV)
리스트는 O(N^2)만큼 시간복잡도
### 왜 이 알고리즘?

### 시간 복잡도
마을 개수 50 이하 => 
도로 정보 2000  -> O(N^2) 이하로 하기!

### 새로 배운 것
1. 힙 쓰려면 리스트로 만들고 그걸 heappush로 만들어야한다!
2. 인접리스트 노드 개수 N알고있다면? 일반 리스트로 만들기!
노드 번호 모르거나 숫자가 아니면 defaultdict로!

```python
import heapq #최소 힙 알
#다음에 방문할 노드 빠르게 찾기!
#거리가 짧은 노드 선택해야 한다
#(거리, 번호))
from collections import defaultdict



def solution(N, road, K):
#N개의 마을 ,K시간 이하로 배달 시간 가중치 가능한 곳만 주문받음
#도로 정보 두 마을 번호, 가중치
    INF = int(1e9)

    distance = [INF] * (N+1)
    visited = [False] * (N+1)
    distance[1] = 0

    graph = [[] for _ in range(N+1)]
    for fro,to,weight in road:
        graph[fro].append((to,weight))
        graph[to].append((fro,weight))

    q = []
    heapq.heappush(q, (0, 1))
    
    while q:
        dist, node = heapq.heappop(q)

        if visited[node] == True:
            continue

        visited[node] = True

        for next_node , weight in graph[node]:
            cost = dist + weight
            if cost < distance[next_node]:
                distance[next_node] = cost
                heapq.heappush(q, (cost, next_node))
                #이 해당 노드까지 가는데 걸리는 거리 = cost

    return sum(1 for d in distance if d <= K)
```
