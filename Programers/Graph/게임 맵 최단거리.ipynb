### 문제 정리
최대한 빨리 도착 
5x5 맵에서 1,1 시작점을 기준으로 5,5까지 도착하는데 
장애물 피해 가는 최단거리

### 왜 이 알고리즘?
최대한 빨리 도착  = bfs
가중치 없음

### 시간 복잡도
맵의 정점은 큐에 최대 1번씩 들어갔다 나오므로
4방향 이동은 상수 시간이 걸리기 때문에
결국 격자 수에 비례합니다
O(N * M)

### 공간 복잡도
move, dist크기 (nxm), q O(n*m)

### 엣지 케이스
1. 벗어나는 범위


### 새로 배운 것
1. deque는 리스트 [] 형태로 값을 넣어줘야 함! 튜플로 넣으면 분해됨! 


```python
from collections import deque

def solution(maps):
    #이동할 수 있는 방향 선언
    move = [[-1,0],[0,-1],[0,1],[1,0]]
    #맵의 크기 지정 변수 선언
    n = len(maps) #세로
    m = len(maps[0])

    #거리 지정 배열을 -1로 초기회
    dist = ([-1] * m for _ in range(n))

        #bfs 함수 선언
    def bfs(start):
        #deque 선언, 시작 위치 추가 
        #큐에서 현재 좌표를 꺼내 4방향으로 다음 위치를 탐색합니다. 
        q = deque([start]) #리스트 형태로 -> dequesㅡㄴ 
        while q: 

        #deque 빌떄까지 반복
            here = q.popleft()
        #현재 위치에서 이동할 수 있는 모든 방향
            for i in move:
                row, column = here[0] + i[0], here[1] + i[1]

            #이동한 위치가, 범위 벗어나면 다음 방향으로 넘어감
            #가드 조건 : 맵 범위, 벽 이동 불가 
                if row < 0 or row >=n or column < 0 or column >= n:
                    continue

                #이동한 위치에 벽이 있는 경우, continue
                if maps[row][column] == 0:
                    continue

                #처음 방문하는 경우, deque 추가하고 거리 계산
                # 방문표시, 거리 업데이트 
                if dist[row][column] == -1:
                    q.append([row, column])
                    dist[row][column] = dist[[here[0]][here[1]]] + 1

                    
        return dict
    
    bfs([0,0])

    #목적지에 닿지 않았다면 초기값 그대로 -1
    return dist[n-1][m-1]
```
