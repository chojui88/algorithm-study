### 문제
n개의 송전탑 = 하나의 트리 형태
하나 끊어서 전략망 2개로 분할하기
둘의 크기를 최대한 맞추려고 한다


### 왜 이 알고리즘?

### 시간 복잡도
O(N x V)
외각 루프 = E만큼 반복
내부 배열 잘라내기 N만큼 반복
dfs 실행 
탐색은 O(N+V)
트리는 O(N)??

### 공간 복잡도
그래프 생성 N만큼
O(N)

### 새로 배운 것

다이타믹 프로그래밍 문제풀때 좋음
2. dfs 돌릴때, 자기 부모로 올라가지 않도록 주의할것!
3. 무방향 트리
4. 간선은 노드부도 하나 더 적음! 반복문 돌릴때 신경쓰기

```python
def build_graph(n, wires):
    graph = [[]for _ in range(n+1)]
    for a, b in wires:
        graph[a].append(b)
        graph[b].append(a)
    return graph

#이번에 끊어진 노드와, 연결된 노드들을 계속계속 dfs 탐색해서 cnt 세기
def dfs(node, parent, graph):
    cnt = 1
    for child in graph[node]:
        if child != parent:
            #방향 거슬러올라가는것 방지! 
            cnt+= dfs(child, node, graph)
    return cnt

def solution(n, wires):
    graph = build_graph(n, wires)
    min_diff = float("inf")

    for a, b in wires:
        graph[a].remove(b)
        graph[b].remove(a)

        cnt_a = dfs(a,b,graph)
        cnt_b = n - cnt_a

        min_diff = min(min_diff,abs(cnt_a - cnt_b))

        graph[a].append(b)
        graph[b].append(a)
    return min_diff
```
