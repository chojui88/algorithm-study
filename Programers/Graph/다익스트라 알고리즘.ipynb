### 문제 정리
가중치 그래프에서 start로부터 모든 노드까지의 단일 출발점 최다경로 구하는 문제
출력값은 각 노드까지 최소 비용을 담은 리스트 

### 왜 이 알고리즘?
다익스트라: 간선의 가중치 양수인 그래프 최단거리
    1이라면 bfs, 가중치 = 우선순위 큐 (최소힙)
### 시간 복잡도
인접 리스트 생성 = 간선 개수만큼, O(E)
최소 비용 노드 찾기 while 정점 개수만큼 수행 (큐)
    이때 pop() 매서드는 O(v logv)
    거리 업데이터 = 간선 수만큼 O(E)
O(E + vlogv)

### 공간 복잡도
인접리스트 O(V+E)
최단거리 저장 distance 배열 O(V)
우선순위 큐 O(V)
O(V+E)

### 엣지 케이스
도달 불가능한 노드 존재ㅣ -1로 변환하여 반환하거나 INF로 둘것인지 확인
자기 자신 처리 distance[start] = 0 초기화 필수!
음수 가중치 : 벨만 포드 사용해야 함
중복 간선: < 구문, 이미 처리된것 continue 처리

### 새로 배운 것
1. 자료구조의 정렬 이해 
heqp 튜플 정렬 , 작은것 먼저
2. 배열 초기화 [1:]
3. 인접 리스트 defaultdict(list)

```python
import heapq #최소 힙 알고리즘을 제공, 가장 적은 비용 찾기, 가장 적은 값이 맨 앞
from collections import defaultdict, deque

INF = 9999999999

#누적 비용, 노드 번호 필요
#비용, 노드번호 순으로 적어야 정렬 가능
def solution(start, numNodes, edges):

    pq = [] #우선순위 큐, 가장 가까운 곳을 바로 뽑아내기

    #(우선순위 기준값, (비용,실제 데이터?)
    heapq.heappush(pq, (0, start))
    #첫 번째 인자로 전달된 리스트에 두번째 원소를 힙규칙에 맞게 추가

    distance = [INF] * (numNodes + 1)
    #아직 방문하지 않은 노드들의 최단거리!
    #노드 번호가 1번부터 numNodes 까지이기떄문에 + 1
    distance[start] = 0
    #우선순위 큐 Heap = 일반 리스트가 힙 역할!!!
    #가장 가까운 노드를 pq에 넣는 코드
    #while 문으로 발생 -> 가장 가까운 노드를 꺼내과, 연결된 이웃중 더 짧은 길을 찾아 qp에 넣기
    
    graph = defaultdict(list)
    #기본값이 빈 리스트인 딕셔너리 만들기
    #본체는 딕셔너리고 내용물이 리스트!
    # 노드 간의 연결정보 메모리에 저장 -> 데이터 형태로 
    for u, v, w in edges:
        graph[u].append((v,w))
    
    while pq: #새로 계산한 cost 
            #힙에서 꺼낸 거리가 이미 최단거리 테이블보다 크다면? 필요 없음
            
        #current_node랑 연결된 다음 노드 바로 찾아야함 -> grph[current] 함 찾아줌
        
        #비용, 실제 
        current_dist, current_node = heapq.heappop(pq)
        #가장 첫번째 값을 꺼내기 - 가장 비용이 적은 튜플을 꺼내고, 힙에서 제거하기
        #언패킹 = 바로 탐색에 활용

        if distance[current_node] < current_dist:
                continue
        
        #더 짧은 경로 -> 새로 계산한 누적비용 cost가 더 작을때! 
        for next_node, weight in graph[current_node]:
            cost = current_dist + weight #현재 노드 - 다음 노드
            #distance 안에 새로운 값 업데이트!  = 노드의 최단거리
            if cost < distance[next_node]:
                distance[next_node] = cost
                heapq.heappush(pq, (cost, next_node)) 

    return distance[1:]
    #start부터 각 도느까지 최소 비용 담은 벡터를 반환
```

```python
import heapq #최소 힙 알고리즘을 제공, 가장 적은 비용 찾기, 가장 적은 값이 맨 앞
from collections import defaultdict, deque

INF = 9999999999

def solution(start, numNodes, edges):
    #그래프 초기화
    graph = defaultdict(list)
    for fron_node, to_node, weight in edges:
        graph[fron_node].append((to_node, weight))

    #최단경로길이, 방문이력 초기화
    distance = [INF] * numNodes
    visited = [False] * numNodes
    distance[start] = 0

    #우선순위 큐 
    priority_queue = [(0,start)]
    #heap호출하지 않아도 됨, 

    while priority_queue:
        current_distance, current_node = heapq.heappop(priority_queue)

        #이미 방문 노드는 무시
        if visited[current_node]:
            continue

        visited[current_node] = True

        #인접노드 거리 업데이트
        for neighbor, weight in graph[current_node]:
            new_distance = distance[current_node] + weight
            if new_distance < distance[neighbor]:
                #neighbor 값 없으므로 INF이다
                distance[neighbor] = new_distance
                heapq.heappush(priority_queue, (new_distance,neighbor))

    return distance
```
