### 문제 파악
시작 노드를 기준으로, 연결된 노드들을 깊이 우선으로 탐색하고
방문한 순서를 반환하는 문제

graph = 간선의 목록
start = 시작 노드

### 왜 이 알고리즘?
graph 값을 인접 리스트로 반환해서
저장하면, 깊이 우선 쉽게할 수 있다

### 시간 복잡도
노듸 10000개 넘지 않는다 => O(NlogN)으로 풀어야 햔다 O(n^2)아슬아슬 
O(V + E) 간선수 + 노드수
노드를 한번씩 방문하고, 간선을 학인한다 

### 공간 복잡도
O(v+e)
인접리스트가 간선을 저장하고 O(E)
visited, result 에도 모든 노드가 들어가니까 o(v)
dfs는 최악의 경우, 깊이가 노드 수 만큼 갈 수 있으므로 

### 엣지 케이스
1. 시작 노드만 있고, 간선이 없으면
.get(node, []) 덕분에 정상적으로 종료된다

2. 이미 방문했던 노드 visited 처리 

### 새로 배운 것
1. from collections import defaultdict
존재하지 않는 키에 접근했을 때 에러를 내지않고 기본값 생성해주는 딕셔너리 
2. 입력된 간선 정보 그대로 사용하는 것보다, 인접 리스트로 변환하면 효율적으로 탐색 가능
3. defaultdict(list) 사용하면, 존재하지 않는 키에도 빈 리스트 자동 생성 가능! 

```python
from collections import defaultdict

adj_list = defaultdict(list)
#키가 없으면 자동으로 빈 리스트를 갖는 딕셔너리를 만들겠다, 어떤 노드들과 연결되어 있는지 
# 그래프를 인접리스트로 반환! 
visited = set()
#그래프에 사이클이 발생할 수 있기 때문에 , set은 특정 노드의 방문 여부를 O(1)에 확인할 수 있기 때문에 
result = []
#전역 변수를 선언! adj_list는 입력받은 그래프를 인접 리스트로 표현


def dfs(node):
    #현재 노드 방문처리 
    visited.add(node)
    #방문 순서 저장
    result.append(node)

    #현재 노드와 연결되어있는 이웃 노드를 하나씩 확인
    for neighbor in adj_list.get(node,[]):
        #예외 방지하고, 간선 없는 외딴 코드 안전하게 처리
        #get 메서드는, 연결된 다음 노드가 없는 노드도 안전하게 처리 가능

        #방문하지 않았다면, DFS 호출해서 더 깊게 탐색 
        if neighbor not in visited:
            dfs(neighbor)

def solution(start, graph):
    #graph는 [출발, 도착]노드 쌍! 리스트
    #입력은 간선의 쌍으로 들어오기 때문에 dfs에서 연결된 노드를 빠르게 찾을 수 있도록
    #인접리스트로 변환하는 것! 
    for u, v in graph:
        adj_list[u].append(v)
        #u의 리스트에 v가 쌓임! 출발 - 도착 

    dfs(start)
    # 모든 노드 깊이 우선 탐색 어떤 순서대로 들어있는 리스트 
    return result
```
