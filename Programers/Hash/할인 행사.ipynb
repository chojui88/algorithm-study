알고리즘: 해시 

새로 알게된 것
1. 두개의 리스트 동시에 순회하려면 zip으로 묶기!
2. 딕셔너리 key가 없을때 += 쓸 수 없음
3. 값이 0인것은 제거를 해줘야 == 연산에서 동일하다고 나옴!
4. del 키삭제

```python
#want = 원하는 제품
def solution(want, number, discount):
    #일정 금액 = 10일간 회원!
    #회원 = 매일 한제품 하나 할인!
    #원하는 제품, 수량이 할인하는 날짜와 10일 연속! 일치할때 맞춰서 회원가입

    #일단 물건 품목 딕셔너리를 만들고
    item = {}
    answer = 0
    for w,n in zip(want,number):
        item[w] = n
    
    #10일차 카운트
    current = {}
    for i in range(10):
        d = discount[i] 
        current[d] = current.get(d,0) + 1
        #값의 수를 1씩 증가

    if current == item:
        answer+=1

    #한 칸씩 밀어가면서 확인하기
    for i in range(10,len(discount)):
        out_item = discount[i-10]
        current[out_item] -=1
        if current[out_item] == 0:
            del current[out_item]

             
        in_item = discount[i] 
        current[in_item] = current.get(in_item,0) + 1
        if current == item:
                answer+=1


    
    #모든 제품을 할인 받을 수 있는 회원 날짜의 총 일수 
    return answer
```
