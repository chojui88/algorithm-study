```python
def solution(arr, target):
    #딕셔너리로 만들어서 arr에 키 입력하고 값은 1로
    seen = {num: 1 for num in arr}
    for key,value in arr:
        if (target - key) in seen:
            return "True"

    return "False"
```

```python
def solution(arr,target):
    hash = [0] * (target + 1)

    for num in arr:
        if num <= target:
            hash[num] = 1

    for num in arr:
        if(num>=target):
            continue
        if((target - num) == num):
            continue
        if (hash[target - num]):
            return True

    return False
```
