```python
from itertools import combinations
#기본 라이브러리, 조합, 순열, 반복 작어베서 많이쓰임
#조합 = 순서 상관없이 뽑기
#순열 permutations 순서 고려해서 
from collections import Counter

def solution(orders, course):
    answer = []

    for c in course: #코스 메뉴 개수, 몇개를 뽑는지
        menu = []
        for order in orders: #모든 주문들 하나하나 뽑아서
            comb = combinations(
                sorted(order), c #orderdㅔ서 c개를 뽑아서 조합을 만듬, 통일하기위해 sort
            )
            menu += comb #메뉴에다가, 조합으로 가능한 모든 수를 추가한다 -> 개수별 가능한 메뉴

        counter = Counter(menu) #딕셔너리 형태로 반환
        if(
            len(counter) !=0 and max(counter.values()) !=1
        ): #가장 많이 주문된 갯수가 2 이상이어야 함
            for m, cnt in counter.items():
                if cnt == max(counter.values()): #가장 많이 주문된 개수
                    answer.append("".join(m))

    return sorted(answer)
```
