알고리즘: 해시 테이블 = 일반 리스트 사용하면, O(n2), 동명이인 처리 카운트 가능
시간복잡도: O(N) = 해시테이블에 추가하는 O(N), 완주 제외도 역시
공간복잡도: O(N) = 딕셔너리에 저장되는 데이터는, 최악의 경우 N개의 쌍을 만들어냄

새로 배운 것
1. dict.get(key, default)
dict[key]로 직접 접근했을때 키가 없으면 에러가 나지만, .get(key, 0)을 쓰면
키가 없을때 기본값을 반환! 
2. dict.items() 암기! 


```python
def solution(participant, completion):
    p_dict = {}
    for name in participant: #이름을 key로 만들고 value에 
        p_dict[name] = p_dict.get(name, 0) + 1
        #딕셔너리에 key 있으면 그거 가져오고, 없으면 0

    for name in completion:
        p_dict[name] -= 1
    
    for name, count in p_dict.items()
        if count == 1:
            return name


    
    
```

```python
def solution(participant, completion):
    dic = {}

    for p in participant:
        if p in dic:
            dic[p] +=1
        else:
            dic[p] = 1

    for c in completion:
        dic[c] -= 1

    for key in dic.keys():
        if dic[key] >0:
            return key
```
