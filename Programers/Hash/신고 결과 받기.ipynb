#테스트케이스
1. 중복 신고 처리 - 같은 사람을 여러번 신고 (set으로 중복 확인)
2. 신고 안당한사람 - count에서 잘 제거 되었는지
3. 모두가 신고당하는 경우
4. 엣지 케이스 - 자기 자신 신고, k = 1
5. 빈 신고 목록 
6. k가 매우 큰 경우 - k = 100
7. 많은 신고 처리 (성능테스트)

```python
def solution(id_list, report, k):
    reported_user = {}
    count = {}
    answer = []
    for re in report:
        user_id, reported_id  = re.split(" ")
        user_id =set(user_id)
        reported_user[reported_id] = reported_user.get(reported_id,set(user_id)) #이거 값에 set어떻게 넣지..?

    for i in id_list:
        count[i] = 0

    for i in count:
        if len(reported_user[i]) >= k:
            count[i] = count.get(i,0)+1

    return count.values()
```

```python
def solution(id_list, report, k):
    reported_user = {}
    count = {}

    for r in report:
        user_id, reported_id = r.split()
        if reported_id not in reported_user:
            reported_user[reported_id] = set()
        reported_user[reported_id].add(user_id)

    for reported_id, user_id_lst in reported_user.items():
        if len(user_id_lst) >= k:
            for uid in user_id_lst:
                if uid not in count:
                    count[uid] = 1
                else:
                    count[uid] +=1
    answer = []
    for i in range(len(id_list)):
        if id_list[i] not in count:
            answer.append(0)
        else:
            answer.append(count[id_list[i]])

    return answer



```
