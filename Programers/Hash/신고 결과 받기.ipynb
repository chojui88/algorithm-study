```python
def solution(id_list, report, k):
    answer = []
    user = {}
    mail = {}
    for id,re in zip(id_list, report):
        user[re] = user[re].get(id,0)+id
    #해쉬로 (신고 당한사람: 신고 한사람) 쌍을 만들음

    for id in id_list:
        mail[id] = 0
    #신고 당한사람이 k번을 넘으면, 신고 한사람에게 모두 메일이 갈 수 있도록
    for i in range(user[re])
        if user[i].count >k#값의 갯수가 :
        #메일 리스트에, 사용자 담고, 그 값에 1씩 더해야지!
            mail[user[i]] +=1 #이거 user안에 값이 여러개인데 하나하나의 maiil키에 1씩 더하려면 어떻게 해야하지?

            
    return mail.value() #각각 유저별 매일 횟수의 값만 출력
```
