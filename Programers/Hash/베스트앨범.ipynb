#새로 배운것

#시간복잡도
O(N log N) 
장르별 총 재생횟수를 기준으로 정렬한 시간복잡도는 O(GlogG)이지만

#공간복잡도
딕셔너리 O(N)


```python
def solution(genres, plays):
    answer = []

    genre_total = {} #장르별 재생횟수
    genre_songs = {} #장르별 (인덱스, 재생횟수)
  
    #일단 장르별로 plays의 총 합을 구해야해 
    for i,(g,p) in enumerate(zip(genres, plays)):
        genre_total[g] = genre_total.get(g,0) + p 
        genre_songs.setdefault(g,[]).append((i,p))
        #default 값으로, 빈 리스트[]생성후 반환

    #키값이, 내림차순으로 value(재생수가 큰 대로) 정렬
    sotred_genre = sorted(genre_total.keys(), key=lambda g: genre_total[g], reverse=True)

    for g in sotred_genre:
        songs = sorted(genre_songs[g], key = lambda x: (-x[1],x[0]))
    #두개의 기준으로 덩시에 정렬 - 내림차순 정렬, 튜플로 묶은 것
        for song in songs[:2]:
            answer.append(song[0])

    return answer
  


```

총 재생 횟수를 기준으로 장르를 내림차순 정렬 ()
각 장르별로 2곡씩 선정해서 플리 만들기
genres는 배열, 인덱스는 노래의 고유번호

```python
def solution(genres, plays):
    answer = []
    genres_dict = {}
    play_dict = {}

    for i in range(len(genres)):
        genre = genres[i]
        play = play[i]
        if genre not in genres_dict: #파이썬은 키가 없을때 초기값을 명시해야함
            genres_dict[genre] = []
            play_dict[genre] = 0
        genres_dict[genre].append((i,play)) #장르 내에서 많이 재생하 곡 뽑을때 사용
        play_dict[genre]+= play #총 재생 횟수

    sorted_genres = sorted(play_dict.items(), key=lambda x: x[1], reverse=True)
    #장르별로, 총 재생홧수가 많은 순으로 정렬이 된다! 

    for genre, _ in sorted_genres: 
        sorted_songs = sorted(genres_dict[genre], key = lambda x: (-x[1],x[0]))
        #정렬된 노래: 장르 속에서, 많이 재생된 노래들이 있는데 거기서 한 장르 키를 기준으로, 정렬
        answer.extend([idx for idx, _ in sorted_songs[:2]])
        #해당 함수를 호출한 리스트에, 인수로 들어온 객체를 이어붙여 그 인덱스만 반환

    return answer
```
