닉네임 변경하면, 모두 수정하면 시간복잡도 O(N2), 100,000이면 쓸 수 없음

시간복잡도: O(N)

공간복잡도:  O(N)
    nickname 딕셔너리 최대 N개

```python
def solution(record):
    #record에서 split해서 3개로 나눠서 {} 만들기
    nickname = {}

    for i in record:
        info = i.split()
        act = info[0]
        
        if act == 'Enter' or act == 'Change':
            key = info[1]
            value = info[2]
            nickname[key] = value

    answer = []
    for i in record:
        info = i.split()
        act = info[0]
        user_id = info[1]

        if act == "Enter":
            answer.append(f"{nickname[user_id]}님이 들어왔습니다.")
        elif act == "Leave":
            answer.append(f"{nickname[user_id]}님이 나갔습니다.")
    
    return answer

    

        
    
```
