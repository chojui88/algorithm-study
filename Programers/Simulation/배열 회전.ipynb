#### 시간 복잡도
1. arr 배열은 2차원 배열
배열을 탐색하고 for문으로 교체하면
O(N^2)

#### 공간 복잡도
1. 2차원 arr 배열 
2. 새로운 rotate 배열

O(n^2)

#### 엣지 케이스

#### 새로 배운 것
1. arr.copy() 원본 배열 복사! 

#### 질문
1. 깊은 복사, 얕은 복사
-> copy() 얕은 복사 : 내부 원소 수정시 원본 영향 있음, 내부 리스트 공유
하지면 함수 자체에서 새 배열을 생성하기 때문에 새값을 채워넣어서 원본 구조가 유지가 됨

-> deepcopy(arr) 깊은 복사 = 이게 안전함! 


```python
#1. arr을 90도로 회전하는 함수 구현하기
def rotate_90(arr):
    # 1-1행의 i ,j 라고 할떄, 한바퀴 돌리면 i -> n-1-j, j -> i
    # 1-2 모든 행열을 바꿔야 하니까 이중 포문으로 모두 바꿔주기 , 이때 원본값 변환시키면 안됨
    n = len(arr)
    rotated_arr = [[0] * n for _ in range(n)]

    for i in n:
        for j in n:
            rotated_arr[i][j] =arr[n-1-j][i]

    return rotated_arr

def solution(arr, n):
    # arr: 2차원 배열 (10 이하)
    # n 회전하는 횟수

    rotated_arr = arr.copy()

    #2. 해당 함수를 n번 수행해서 나오는 결과값 리턴
    for _ in range(n):
        rotated_arr = rotate_90(rotated_arr)

    return rotated_arr 
    #90도로 n번 회전하는 함수
```
