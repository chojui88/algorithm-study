#문제
1. 메모리 초과
-> n의 크기만큼 리스트 공간을 만들어놓기 떄문에!
지금 n의 가능 범위는 10^12 이다. 
리스트 하나당 8 byte -> 백준 메모리 제한은 보통 128, 256MB -> 100만? 10^7 정도
-> 
#새로 배운 것
1. 띄어쓰기로 정수 받는것 => map으로 정수 받고 이를 split으로 나눠주야 함
2. 빈 리스트는 인덱스로 값을 접근할수 없다 => 그래서 처음에 크기를 정해주어야 함


```python
#무한 수열
#A0 = 1
#A[i] = A[i//p] + A[i//Q] 


n, p, q = map(int, input().split())

A = {}
A[0] = 1
#p<q
#A[p] = A[1] + A[p//q]


for i in range(1, n+1):
    #A[8//4]
    #a[2] = a[2//2] + a[2//3] = a[1] + a[0]
    #a[4]= a[4//2] + a[]
    A[i] = A[i//p] + A[i//q]

answer = A[n]
print(answer)
#7 / 2 = 3.5
#7 / 3 = 2.33333
#즉 A[6]
#A의 n번째 수열 값 출력

```

```python
#시간 초과 문제 = 탑다운 방식으로 해결
n, p, q = map(int, input().split())

A = {}
A[0] = 1
#p<q
#A[p] = A[1] + A[p//q]

#재귀함수로 풀기!
def dp(i) :
    #만약 A에 있다면 그거 반환하고
    if i in A:
        return A[i]

    #그렇지 않다면 A[i] 를 dp 재귀함수로 계산해서 (계산할때만 쓰기 -> 메모리 절약)
    #A값 반환하기
    A[i] = dp(i//p) + dp(i//q)
    return A[i]


print(dp(n))
#7 / 2 = 3.5
#7 / 3 = 2.33333
#즉 A[6]
#A의 n번째 수열 값 출력
```
