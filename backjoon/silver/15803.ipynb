```python
#스쿼드 4명, 
#준서 저격, 좌표 없음

#x,y 각 팀원의 위치 
#100명의 플레이어, 한 명의 승자가 남을때까지
# 3 명이 직선으로 있으면 동시에 쓰러뜨릴 수 있음 
#각자 직선이 되지 않도록 포지션 잡기 
onex,oney = map(int(input().split()))
twox, twoy = map(int(input().split()))
thrx, thry = map(int(input().split()))
#직선이 되는지 판별이 되는 기계 

#1. 각 좌표의 x 나 y 값이 같거나
if onex == twox == thrx:
    print("WHERE IS MY CHICKEN?")

elif oney == twoy == thry:
    print("WHERE IS MY CHICKEN?")

#2. 1차 함수로 직선을 만들었을때, 그 좌표에 일치하는 직선

else: 
#y = nx + b 

    # 기울기 n 구하기
    n = (twoy - oney) / (twox - onex)

    # y절편 b 
    b = oney - n * onex

    if abs(thry - (n * thrx + b)) < 9999999:
        print("WHERE IS MY CHICKEN?")
    else:
        print('WINNER WINNER CHICKEN DINNER!')
```
