```python
#스트라이트 x, 타자가 공 휘두룸x
n = int(input())
# 창석이가 던진 공의 수
lst = list(map(int,input().split()))
#볼, 몸에 맞는 공, 폭투 
# 1 2 3
#1, 3 둘다 볼!
base = [False,False,False]
base_point = -1
ball_point = 0
answer = 0
#만약 2번이 있다면? 바로 그냥 냅다 빠따 던지고 진루
for i in lst:
    if i == 2 :
        base_point +=1
        if base_point == 3:
            answer+=1
            base_point = 3
        else:
            base[base_point] = True
    elif i == 1 or i == 3:
        ball_point+=1
        if ball_point == 4:
            base_point+=1
            if base_point == 3:
                answer+=1
                base_point = 3
            else: 
                base[base_point] = True
                ball_point = 0


print(answer)
    
        

#1 x  4번 => 타자는 1루
#2 => 타자는 1루

#2루는 
#2.1 1루에 사람이 있으면 ? 3루
#2.1 없으면? 그자리

#3루는
#3.1 1,2,루에 사람이 있으면? 홈 + 1득점!
#3.2 없으면? 그자리


#총 실점 구하기
```
