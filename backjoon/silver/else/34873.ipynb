1. counter이 무슨 라이브러리야
-> collectios 표준 라이브러리에 포함된 클래스!
리스트나 문자열에서 각 원소가 몇번 등장하느냐! 
counter(a) -> 원소는 key로 , 등장 횟수는 value로 저장한다.
즉 a를 counter 딕셔너리로 만들겠다는 뜻!!
count[1] = 등장 횟수 (1이 key 니까)
count.keys() 키의 종류들
count.values() 각각 키의 값이 몇이냐 (키별 등장 횟수가 몇이나)
count.items() 원소와 등장 횟수, 키와 밸류 둘다 인출! 
max(cnt.values()) 가장 value 값이 큰 키의 값은 무엇이냐!
count.most_common(1) = 가장 많이 등장한 원소와 횟수? .. 모르겠다 이근


```python
from collections import Counter

#입력 : N
n = int(input())
#사탕의 종류 2 x n 개의 정수
a = list(map(int, input().split()))
#각각 종류별 map에 넣어서
count = Counter(a)

#가장 많은 종류의 개수가 2 이하여야 , 2개를 나눴을때 모두 다른 종류를 가질 수 
if max(count.values()) <= 2:
    print("Yes")
else:
    print("No")

#친구들과 서로 모두 다른 종류의 사탕 n개 
#yes, 아니면 no

#서로 다른 사탕의 종류가 몇개?
```
