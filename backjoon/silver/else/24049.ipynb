#질문
1. map 이라는 함수명은 쓰면 안되곘지 ? graph는 쓸 수있나?
2. graph = [[]* m for _ in range(n)] 이게 n x m 이 맞나?

```python
#세로n 가로 m 
n, m = map(int, input().split())

#왼쪽 가장자리 꽃 n개의 정보가   i 번째 수는 정원의 i행 왼쪽, 
#세로가 0
graph = [[-1]* (m+1) for _ in range(n+1)]

left = list(map(int, input().split()))
top = list(map(int, input().split()))

for i in range(1,m+1): #세로 길이만큼 받아야m=
    graph[0][i] = top[i-1]

for j in range(1,n+1): 
    graph[j][0] = left[j-1]

#위쪽 가장자리 꽃 m개 
#각각 왼쪽, 위쪽 가장자릴 0,0은 비어있다고 하고 

#1.1 부터 채워가기 bfs? 
for j in range(1,n+1): #세로
    for i in range(1,m+1): #가로
        if graph[j][i] == -1:
            if (graph[j-1][i] + graph[j][i-1]) %2 == 0: #같다
                graph[j][i] = 0
            else:
                graph[j][i] = 1
            
print(graph[n][m])


#세로 번호가 작은 칸부터, 같으면 열 번호 작은 칸부터 심는데 (세로 먼저 쭉 그다음 가로)
#왼쪽, 위쪽 꽃을 보고 두 색이 같으면 노란, 다르면 빨강

#행 m열에 해당하는 칸에 어떤 꽃을 심게 될지 궁금해졌다. 이를 알아내는 프로그램 작성


```
